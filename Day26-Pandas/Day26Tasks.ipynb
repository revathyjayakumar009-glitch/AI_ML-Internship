**Task 2: Series Practice** 
1. Create Series  
2. Create Series with custom index  
3. Access Series values  

```python
#Create Series
numbers = pd.Series([10, 20, 30, 40])
print(numbers)
```

```python
#create series with custom index
students = pd.Series(
    [85, 90, 78],
    index=["Anu", "Rahul", "Meera"]
)

print(students)
```

```python
#Access series values
print(students["Rahul"])
```

**Task 3: DataFrame Practice**
1. Create DataFrame using dictionary  
2. Print DataFrame  
3. Add one more column 

```python
#Create Dataframe using dictionary
data = { 
         "Subject":["Maths","IT","DA"],
         "Marks":[95,90,85]
}

df = pd.DataFrame(data)
```

```python
#Print DataFrame
df
```

```python
#Add one more column
df["Grade"] = ["AA","A","A"]
```

```python
df
```
