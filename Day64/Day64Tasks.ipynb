# Practical Task 2 
Write a Python program that sends a greeting to the AI model and displays the response. 

```python
pip install groq python-dotenv
```

```python
from dotenv import load_dotenv
import os
from groq import Groq
```

```python


# Load environment variables from .env
load_dotenv()

# Read the API key
api_key = os.getenv("GROQ_API_KEY")

# Create the Groq client
client = Groq(api_key=api_key)

# Send a greeting to the AI model
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": "Hello! How are you today?"
        }
    ]
)

# Display the AI response
print("AI Response:")
print(response.choices[0].message.content)
```

```python
# Load environment variables
load_dotenv()

# Read API key from .env
api_key = os.getenv("GROQ_API_KEY")

# Create Groq client
client = Groq(api_key=api_key)

print("=" * 50)
print("🤖 Welcome to the AI Chatbot")
print("Type 'bye' to end the conversation.")
print("=" * 50)

while True:
    # Get user input
    user_input = input("You: ")

    # Exit condition
    if user_input.lower() == "bye":
        print("Bot: Goodbye! Have a great day! 👋")
        break

    # Send user message to the AI model
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": "You are a friendly AI assistant."
            },
            {
                "role": "user",
                "content": user_input
            }
        ]
    )

    # Display AI response
    print("Bot:", response.choices[0].message.content)
```

# **Practical Task 4** 
Create a chatbot that answers questions related to: 
• Artificial Intelligence  
• Python  
• Machine Learning  
• Data Science 

```python
# Load API key from .env file
load_dotenv()
api_key = os.getenv("GROQ_API_KEY")

# Create Groq client
client = Groq(api_key=api_key)

print("=" * 60)
print("🤖 AI Learning Chatbot")
print("=" * 60)
print("Ask questions about:")
print("• Artificial Intelligence")
print("• Python")
print("• Machine Learning")
print("• Data Science")
print("Type 'bye' to exit.\n")

while True:
    user_input = input("You: ")

    # Exit condition
    if user_input.lower() == "bye":
        print("Bot: Goodbye! Happy Learning! 👋")
        break

    # Send request to Groq AI
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an educational AI chatbot. "
                    "Answer only questions related to Artificial Intelligence, "
                    "Python, Machine Learning, and Data Science. "
                    "If the user asks about any other topic, reply politely: "
                    "'Sorry, I can only answer questions related to Artificial Intelligence, Python, Machine Learning, and Data Science.'"
                )
            },
            {
                "role": "user",
                "content": user_input
            }
        ]
    )

    print("\nBot:", response.choices[0].message.content)
    print()
```
