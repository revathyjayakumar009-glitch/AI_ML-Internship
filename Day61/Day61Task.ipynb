Convert the following sentence to lowercase: <br>

# **Practical Task 5**
Create a table showing the difference between Stemming and Lemmatization.

| **Feature**    | **Stemming**                                                                       | **Lemmatization**                                                        |
| -------------- | ---------------------------------------------------------------------------------- | ------------------------------------------------------------------------ |
| **Definition** | Removes prefixes or suffixes to obtain the root form of a word using simple rules. | Converts a word to its dictionary (base) form using linguistic analysis. |
| **Output**     | May produce non-dictionary words.                                                  | Produces valid dictionary words.                                         |
| **Accuracy**   | Less accurate.                                                                     | More accurate.                                                           |
| **Speed**      | Faster.                                                                            | Slower.                                                                  |
| **Context**    | Does not consider the meaning or context of the word.                              | Considers the meaning and context of the word.                           |
| **Example**    | *Studies* → **Studi**                                                              | *Studies* → **Study**                                                    |
| **Example**    | *Running* → **Run**                                                                | *Running* → **Run**                                                      |
| **Example**    | *Better* → **Better**                                                              | *Better* → **Good**                                                      |
