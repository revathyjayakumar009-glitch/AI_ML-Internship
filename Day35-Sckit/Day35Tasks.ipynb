```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
#Avoid Warnings
import warnings
warnings.filterwarnings('ignore')
```

# **Task 2: Dataset Practice** 
1. Create student dataset  
2. Create salary dataset  
3. Create sales dataset 

```python
#Create Student dataset

student_data = {
    
    'Study_Hours':[1,2,3,4,5,6,7,8],
    'Marks':[20,30,40,50,60,70,80,90]

}

student_df = pd.DataFrame(student_data)

student_df
```

```python
#Create Salary dataset
salary_data =  {
   
    'Experience': [1,2,3,4,5],
    'Salary': [25000,35000,45000,55000,65000]
}

salary_df = pd.DataFrame(salary_data)

salary_df
```

```python
#create sales dataset
sales_data = {
    'Month_Number': [1,2,3,4,5],
    'Sales': [20000,25000,30000,35000,40000]
}

sales_df = pd.DataFrame(sales_data)

sales_df
```

# **Task 3: Model Training** 
1. Train linear regression model  
2. Use fit()  
3. Understand learning process 

# Train linear regression model
A Linear Regression model is used to predict numerical values based on input data.

Example:

Study hours → Marks
Experience → Salary

# Training student dataset

```python
# Features and Label
X = student_df[['Study_Hours']]  # Input/Feature data or Independent variable
y = student_df[['Marks']]        #Output/Label or Dependent variable

#Create model
model = LinearRegression()

#Usefit
model.fit(X,y)    #fit() function is used to train machine learning model

print("Model Training completed")
```

#### 3. Understand Learning Process
How the Model Learns
##### Step 1: Input Data

The model receives:

Features (study hours)
Labels (marks)
#### Step 2: Pattern Finding

The model identifies relationships.

Example:

More study hours → Higher marks
#### Step 3: Create Mathematical Relationship

Linear Regression learns a straight-line relationship:
y=mx+b
Where:

y = predicted output
x = input feature
m = slope

#### Step 4: Prediction

After learning, the model predicts outputs for new data.

# Training the salary dataset

```python

#Feature and Label
X= salary_df[["Experience"]]
y= salary_df[["Salary"]]

#Train Model
model = LinearRegression()
#Use fit
model.fit(X,y)

print("Model training completed")
```

# Training Sales values

```python
#Features and Label
X= sales_df[["Month_Number"]]
y= sales_df[["Sales"]]

#Train Model
model =  LinearRegression()
#Use fit
model.fit(X,y)

print("Model training completed")
```

# **Task 4: Prediction Practice**

# Explain Training Logic
Training means teaching the ML model using data.
### Step 1: Give Input Data

The model receives features (X).<br>

**Example:** <br>

* Study hours
### Step 2: Give Correct Output

The model receives labels (y).<br>

**Example:** <br>

* Marks
### Step 3: Learn Relationship

The model identifies patterns between input and output.<br>

**Example:** <br>

* Higher study hours → Higher marks

* Linear Regression learns a relationship:
  y=mx+b
Where:<br>

y = output<br>
x = input<br>
m = slope<br>
b = intercept<br>
### Step 5: Predict New Values

Study Hours = 7<br>
