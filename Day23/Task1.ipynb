# 1. What is broadcasting?  

Broadcasting is a feature in NumPy that allows arrays with different shapes to perform arithmetic operations automatically without manually resizing them.

It helps NumPy:

1. Avoid loops
2. Write shorter code
3. Improve performance

```python
import numpy as np
```

```python
#Simple broad casting
arr = np.array([1,2,3])
print(arr + 5)
```

# 2. Why broadcasting is important? 

#### 1. Faster Computation

NumPy performs operations internally using optimized C code, making broadcasting much faster than Python loops.

#### 2. Less Code

Broadcasting reduces lengthy loop-based code into simple one-line operations.

#### 3. Memory Efficient

NumPy does not create unnecessary copies of data while broadcasting.
It stretches values logically, saving memory.

#### 4. Easy Matrix Operations

Useful for:

1. Adding rows or columns
2. Scaling datasets
3. Mathematical transformations
4. Machine learning calculations

#### 5. Important in AI & Data Science

Broadcasting is heavily used in:

1. Neural networks
2. Feature scaling
3. Image processing
4. Tensor operations
5. Deep learning frameworks

# 3. What are broadcasting rules? 

#### Rule 1: Dimensions Must Be Equal

If dimensions are the same, broadcasting works.

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([[10,20,30],
              [40,50,60]])

print(a + b)
```

```python
print(a.shape)
print(b.shape)
```

#### Rule 2: One Dimension Can Be 1

If one dimension is 1, NumPy stretches it automatically.

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a + b)
```

#### Rule 3: Missing Dimensions Are Treated as 1

Smaller arrays are padded with 1 on the left side.

#### Rule 4: If Dimensions Are Not Compatible → Error

Broadcasting fails when:

Dimensions are different
AND
Neither dimension is 1

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([1,2])

print(a + b)
```

# 4. What happens with incompatible shapes? 

When two arrays have incompatible shapes, NumPy cannot perform broadcasting.
