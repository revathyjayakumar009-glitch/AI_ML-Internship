# 1. What is correlation? 

Correlation is a statistical method used to measure the relationship between two variables.

###### It shows:

How strongly two variables are related
Whether they move in the same direction or opposite directions
#### Types of Correlation
##### Positive Correlation (+)
Both variables increase together
Example: Study hours ↑ → Marks ↑
##### Negative Correlation (-)
One variable increases while the other decreases
Example: Screen time ↑ → Sleep ↓
##### No Correlation (0)
No relationship between variables
Example: Shoe size and exam marks
##### Correlation Value Range
+1 → Perfect positive correlation
0 → No correlation
-1 → Perfect negative correlation

# 2. Difference between positive and negative correlation  

| Positive Correlation                                | Negative Correlation                           |
| --------------------------------------------------- | ---------------------------------------------- |
| Both variables move in the same direction           | Variables move in opposite directions          |
| If one variable increases, the other also increases | If one variable increases, the other decreases |
| Correlation value is between `0` and `+1`           | Correlation value is between `0` and `-1`      |
| Example: Study hours and marks                      | Example: Exercise time and body weight         |
| Graph trend goes upward 📈                          | Graph trend goes downward 📉                   |


# 3. What is heatmap?  

A heatmap is a graphical representation of data where values are shown using different colors.

In Data Science, heatmaps are mainly used to visualize:

* Correlation between variables
* Patterns in datasets
* High and low values easily

Different colors represent different values:

* Dark/strong colors → higher values
* Light/soft colors → lower values
* Heatmap in Seaborn

Seaborn provides the heatmap() function to create heatmaps easily.

**Syntax**<br>
sns.heatmap(data)

# 4. What is pairplot? 

A pairplot is a Seaborn visualization used to compare relationships between multiple numerical columns in a dataset.

It creates:

Scatter plots between columns
Histograms/Distribution plots on the diagonal

Pairplots help in:

* Understanding relationships between variables
* Finding patterns and trends
* Detecting correlations and outliers<br>
**Syntax**<br>
  sns.pairplot(df)

# 5. Why correlation is important in ML?  

Correlation is important in Machine Learning because it helps understand the relationship between features (variables) in a dataset.

It is mainly used during:

Exploratory Data Analysis (EDA)
Feature Selection
Data Preprocessing
Importance of Correlation in ML
**1. Feature Selection**

Correlation helps identify which features are strongly related to the target variable.

Strong correlation → useful feature
Weak correlation → less useful feature

This improves model performance.

**2. Removes Redundant Features**

Sometimes two features contain almost the same information.

Example:

Salary and Annual Income
Height in cm and Height in meters

Highly correlated features can create redundancy.

Removing them helps:

Reduce complexity<br>
Improve model efficiency<br>
**3. Improves Model Accuracy**

Using relevant correlated features helps the model learn patterns better and make accurate predictions.

**4. Helps Detect Relationships**

Correlation helps understand:

* Positive relationships
* Negative relationships
* No relationships

This gives better insights into the dataset.

**5. Reduces Overfitting**

Removing unnecessary highly correlated features can reduce overfitting and improve generalization.

**Example**

If study hours and marks have high positive correlation, a Machine Learning model can use study hours to predict marks effectively.
