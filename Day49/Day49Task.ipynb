```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
```

```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
```

```python
results = model.train(
    data=r"D:\AI-ML Internship\Day49\Person detection.yolov8\data.yaml",
    epochs=10,
    imgsz=320,
    batch=2,
    workers=0,
    device="cpu"
)
```

## Practical Task 1 – Observation of YOLO Training
### 1. Epoch
The model was trained for 10 epochs.<br>
Epochs increased from 1 to 10.<br>

Observation: The model completed all training cycles successfully.<br>
### 2. Loss
| Epoch | Box Loss | Cls Loss | DFL Loss |
| ----- | -------- | -------- | -------- |
| 1     | 2.909    | 3.541    | 2.193    |

### Observation:

* Box Loss decreased from 2.909 → 1.979

* DFL Loss decreased from 2.193 → 1.737

✅ Loss values decreased, showing that the model learned from the dataset.

### Observation:

### 4. Recall

### Observation:

* Recall remained 100%.

✅ No objects were missed during validation.
### 5. mAP
| Epoch | mAP50 | mAP50-95 |
| ----- | ----- | -------- |

### Observation:
