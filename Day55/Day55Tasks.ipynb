#  **Practical Task 3**
Print vocabulary using:<br> 
get_feature_names_out()<br>

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
documents = [ 
"AI is amazing", 
"AI is powerful", 
"Machine learning is amazing" 
] 
vectorizer = TfidfVectorizer() 
X = vectorizer.fit_transform(documents) 
# Convert to DataFrame
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
print("\nTF-IDF Matrix:")
print(df)
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

```

# **Practical Task 4** 
Compare BoW and TF-IDF outputs.

```python
bow = CountVectorizer()
bow_matrix = bow.fit_transform(documents)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow.get_feature_names_out()
)

print("Bag of Words (BoW):")
print(bow_df)


tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\nTF-IDF:")
print(tfidf_df.round(3))
```

# **Practical Task 5**
Identify the most important words using TF-IDF scores.

```python
print("\nMost Important Word in Each Document:\n")

for i, row in df.iterrows():
    important_word = row.idxmax()      # Word with highest TF-IDF score
    score = row.max()                  # Highest TF-IDF score
    print(f"Document {i+1}: {important_word} ({score:.3f})")
```
