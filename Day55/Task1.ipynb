# 1. What is TF-IDF?  


TF-IDF (Term Frequency–Inverse Document Frequency) is a text representation technique used in Natural Language Processing (NLP) to measure how important a word is in a document compared to a collection of documents (called a corpus).

* Term Frequency (TF): Measures how often a word appears in a document.
* Inverse Document Frequency (IDF): Measures how unique or rare the word is across all documents.

The TF-IDF score is calculated as:

TF-IDF=TF×IDF

Words that appear frequently in one document but rarely in other documents receive high TF-IDF scores, making them more important for tasks such as document classification, information retrieval, and search engines.

# 2. Define TF. 

Term Frequency (TF) is a measure of how often a word (term) appears in a document. It indicates the importance of a word within that specific document.
#### Formula:
TF=Number of times a word appears in a document / Total number of words in the document
#### Example:
Document: "AI is powerful and AI is amazing"

* Number of times "AI" appears = 2
* Total number of words = 7
  TF(AI)=2 / 7 =0.286<br>
Answer: The Term Frequency (TF) of "AI" is 0.286.

# 3. Define IDF. 

Inverse Document Frequency (IDF) measures how unique or rare a word is across a collection of documents (corpus). Words that appear in many documents receive a lower IDF, while words that appear in only a few documents receive a higher IDF.
#### Formula:
         IDF=  log (Total number of documents / Total number of documents )   
#### Example:

Consider these three documents:

1. AI is powerful
2. AI is amazing
3. Machine Learning is powerful
For the word "AI":

* Total number of documents = 3
* Number of documents containing "AI" = 2<br>
IDF(AI)  = log (3 / 2 ) = 0.176

# 4. Why is IDF important? 

Inverse Document Frequency (IDF) is important because it helps identify how unique or informative a word is across a collection of documents. It reduces the importance of common words that appear in many documents and increases the importance of rare words that carry more meaningful information.

#### Why IDF is important:

* Reduces the weight of common words (e.g., is, the, and).
* Gives higher importance to rare and meaningful words.
* Improves the accuracy of text classification and information retrieval.
* Helps search engines and NLP models identify the most relevant keywords.
* Works with TF to produce more informative TF-IDF scores.

| **Bag of Words (BoW)**                        | **TF-IDF**                                                                    |
| --------------------------------------------- | ----------------------------------------------------------------------------- |
| Counts the frequency of words in a document.  | Assigns weights based on word importance.                                     |
| Gives equal importance to all words.          | Gives higher importance to unique words and lower importance to common words. |
| Common words may dominate the feature vector. | Reduces the influence of common words.                                        |
| Simple and easy to implement.                 | Slightly more complex due to TF and IDF calculations.                         |
| Suitable for basic text analysis.             | Better for text classification, search engines, and information retrieval.    |
