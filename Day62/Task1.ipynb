# 1. Define NLTK. .

NLTK (Natural Language Toolkit) is a popular open-source Python library used for Natural Language Processing (NLP). It provides tools for tasks such as tokenization, stop word removal, stemming, lemmatization, part-of-speech tagging, and text classification, helping computers understand and process human language.

# 2. Explain Tokenization with an example.  

Tokenization is the process of breaking a piece of text into smaller units called tokens, such as words or sentences. It is one of the first and most important steps in Natural Language Processing (NLP).
* Example:

* Input Sentence:<br>
Artificial Intelligence is transforming education.<br>
* Output (Word Tokens):<br>
['Artificial', 'Intelligence', 'is', 'transforming', 'education', '.']

# 3. Explain Stop Word Removal.  

Stop Word Removal is the process of removing common words (called stop words) from a sentence that do not add significant meaning to the text. Examples include words like "is," "am," "the," "and," "of," and "to." Removing these words helps NLP models focus on the important keywords.
* Example:

Input Sentence:<br>
I am learning Natural Language Processing using Python.<br>
After Stop Word Removal:<br>
['learning', 'Natural', 'Language', 'Processing', 'using', 'Python', '.']

# 4. Differentiate Stemming and Lemmatization. 

| **Stemming**                                                 | **Lemmatization**                                             |
| ------------------------------------------------------------ | ------------------------------------------------------------- |
| Reduces a word to its root by removing prefixes or suffixes. | Converts a word to its dictionary (base) form called a lemma. |
| Uses simple rule-based methods.                              | Uses vocabulary and linguistic analysis.                      |
| May produce words that are not valid English words.          | Produces meaningful and valid words.                          |
| Faster and less accurate.                                    | Slower but more accurate.                                     |
| Example: **running → run**, **studies → studi**              | Example: **running → run**, **studies → study**               |


# 5. Compare Rule-Based and NLP-Based Chatbots.  

| **Rule-Based Chatbot**                                       | **NLP-Based Chatbot**                                            |
| ------------------------------------------------------------ | ---------------------------------------------------------------- |
| Works using predefined rules and keywords.                   | Uses Natural Language Processing (NLP) to understand user input. |
| Responds only to specific commands or patterns.              | Understands the meaning and context of user queries.             |
| Limited flexibility and cannot handle complex conversations. | Can manage more natural and varied conversations.                |
| Easier to build and suitable for simple tasks.               | More advanced and requires NLP libraries or AI models.           |
| **Example:** Menu-driven college enquiry chatbot.            | **Example:** ChatGPT, Google Assistant, Siri.                    |
