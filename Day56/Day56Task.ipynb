```python
from gensim.models import Word2Vec 
```

# **Practical Task 2** 
Create a small dataset with 5 sentences.

```python
sentences = [
    ["ai", "is", "transforming", "the", "world"],
    ["machine", "learning", "is", "a", "part", "of", "ai"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["python", "is", "popular", "for", "data", "science"],
    ["artificial", "intelligence", "improves", "automation"]
]

print(sentences)
```

# **Practical Task 3**
Train a Word2Vec model. 

```python
model = Word2Vec( 
    sentences, 
    vector_size=50, 
    window=3, 
    min_count=1 
)
```

# **Practical Task 4**
Print word vectors.

```python
print( 
    model.wv["ai"] 
) 
```

# **Practical Task 5**
Find similar words using: 
most_similar()

```python
print( 
model.wv.most_similar("ai") 
)
```
