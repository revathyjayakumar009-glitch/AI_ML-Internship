# **Practical Task 1** 
Create a College Enquiry Menu Chatbot with the following options: <br>
• Course Fee  
• Duration  
• Eligibility  
• Location  
• Contact  
• Exit 

```python
while True: 
 
    print("\n============================") 
    print("    College Enquiry Chatbot") 
    print("============================") 
    print("1. Course Fee") 
    print("2. Course Duration") 
    print("3. Eligibility") 
    print("4. College Location") 
    print("5. Contact Number") 
    print("6. Exit") 
 
    choice = input("\nEnter your choice: ") 
 
    if choice == "1": 
        print("Course Fee: ₹15,000") 
 
    elif choice == "2": 
        print("Duration: 3 Months") 
 
    elif choice == "3": 
        print("Eligibility: Plus Two Pass") 
 
    elif choice == "4": 
        print("Location: Kochi") 
 
    elif choice == "5": 
        print("Phone: 9876543210") 
 
    elif choice == "6": 
        print("Thank You for Visiting!") 
        break 
 
    else: 
        print("Invalid Choice! Please try again.")
```

# **Practical Task 5** 
Modify the chatbot to ask: <br>
Would you like to continue? (yes/no) <br>
If the user enters no, the chatbot should exit gracefully.

```python
responses = {
    "1": "Course Fee: ₹15,000",
    "2": "Duration: 3 Months",
    "3": "Eligibility: Plus Two Pass",
    "4": "Location: Kochi",
    "5": "Phone: 9876543210",
    "6": "Hostel Information:\n- Separate hostels for boys and girls.\n- Wi-Fi, Mess, and 24/7 Security available.",
    "7": "Placement Information:\n- Placement Assistance Available.\n- Top Recruiters: TCS, Infosys, Wipro, Accenture."
}

while True:

    print("\n============================")
    print("    College Enquiry Chatbot")
    print("============================")
    print("1. Course Fee")
    print("2. Course Duration")
    print("3. Eligibility")
    print("4. College Location")
    print("5. Contact Number")
    print("6. Hostel Information")
    print("7. Placement Information")
    print("8. Exit")

    choice = input("\nEnter your choice (1-8): ")

    if choice == "8":
        print("Thank You for Visiting!")
        break

    elif choice in responses:
        print("\n" + responses[choice])

    else:
        print("\nInvalid Choice! Please try again.")

    continue_choice = input("\nWould you like to continue? (yes/no): ").lower()

    if continue_choice == "no":
        print("Thank You for Visiting! Have a Nice Day!")
        break
    elif continue_choice != "yes":
        print("Invalid input! Returning to the main menu.")
```
