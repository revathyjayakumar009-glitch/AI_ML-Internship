# 1. What is SVM? 

Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks. It works by finding the best decision boundary (hyperplane) that separates data points of different classes with the maximum possible margin.

For example, in a pass/fail prediction problem, SVM finds the optimal line (or hyperplane) that separates students who pass from those who fail.

# 2. What is a Decision Boundary? 

A Decision Boundary is a line, curve, or hyperplane that separates different classes in a classification problem.

In SVM, the decision boundary is the optimal hyperplane that divides the data into different categories while maximizing the margin between classes.

Example:

Suppose we have two classes:

* Class A: Pass
* Class B: Fail

The decision boundary is the line that separates the "Pass" data points from the "Fail" data points.

# 3. What are Support Vectors?  

Support Vectors are the data points that lie closest to the decision boundary (hyperplane) in an SVM model.

These points are the most important observations because they determine the position and orientation of the decision boundary. If the support vectors change, the hyperplane may also change.

Example:

In a Pass/Fail classification problem, the students whose scores are closest to the separating boundary between "Pass" and "Fail" are the support vectors.

# 4. What is Margin?  

Margin is the distance between the decision boundary (hyperplane) and the nearest data points (support vectors) from each class.

In SVM, the goal is to find a hyperplane with the maximum margin, which helps improve the model's ability to classify new data correctly.

Example:

If a line separates "Pass" and "Fail" students, the margin is the gap between the line and the closest students from both classes.

# 5. Advantages of SVM? 

1. Effective in High-Dimensional Data<br>
Works well when there are many features or variables.
2. Good Classification Accuracy<br>
Often provides high accuracy for classification tasks.
3. Handles Non-Linear Data<br>
Uses kernel functions (RBF, Polynomial, etc.) to classify complex datasets.
4. Less Overfitting<br>
Maximizing the margin helps improve generalization on new data.
5. Works Well with Small and Medium Datasets<br>
Can perform effectively even when the dataset is not very large.
6. Uses Only Important Data Points<br>
Depends mainly on support vectors, making it memory efficient.
