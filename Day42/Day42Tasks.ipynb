## **Student Result Dataset**

```python
student_data = pd.DataFrame({
    'Study_Hours': [8, 6, 4, 2, 1],
    'Attendance': [90, 80, 70, 50, 40],
    'Result': ['Pass', 'Pass', 'Pass', 'Fail', 'Fail']
})
# Features and Target
X1 = student_data[['Study_Hours', 'Attendance']]
y1 = student_data['Result']

# Create Model
model1 = GaussianNB()

# Train Model
model1.fit(X1, y1)

# Prediction
prediction1 = model1.predict([[5, 75]])

print("Predicted Result:", prediction1[0])
```

```python
# Prediction practice
student_prediction = model1.predict([[2, 55]])

print("Predicted Student Result:", student_prediction[0])
```

## **Disease Dataset**

```python
disease_data = pd.DataFrame({
    'Age': [25, 30, 22, 40, 28],
    'Fever': [102, 101, 98, 103, 97],
    'Disease': ['Yes', 'Yes', 'No', 'Yes', 'No']
})
# Features and Target
X2 = disease_data[['Age', 'Fever']]
y2 = disease_data['Disease']

# Create Model
model2 = GaussianNB()

# Train Model
model2.fit(X2, y2)

# Prediction
prediction2 = model2.predict([[35, 102]])

print("Disease Prediction:", prediction2[0])
```

```python
# Prediction Practice
disease_prediction = model2.predict([[24, 98]])

print("Predicted Disease Status:", disease_prediction[0])
```

## **Spam Dataset**

```python
spam_data = pd.DataFrame({
    'Free_Word': [1, 1, 0, 0, 1],
    'Offer_Word': [1, 0, 1, 0, 1],
    'Spam': ['Yes', 'Yes', 'No', 'No', 'Yes']
})
# Features and Target
X3 = spam_data[['Free_Word', 'Offer_Word']]
y3 = spam_data['Spam']

# Create Model
model3 = GaussianNB()

# Train Model
model3.fit(X3, y3)

# Prediction
prediction3 = model3.predict([[1, 1]])

print("Spam Prediction:", prediction3[0])
```

```python
# Prediction Practice
spam_prediction = model3.predict([[0, 0]])

print("Predicted Spam Status:", spam_prediction[0])
```

# **Task 5: Probability Activity**
Suppose: 
• Total Emails = 200  
• Spam Emails = 150  
Calculate: 
1. Probability of Spam  
2. Probability of Not Spam 

Given:
Total Emails = 200
Spam Emails = 150
### 1. Probability of Spam

##### Formula:

P(Spam)=Number of Spam Emails/Total Emails <br>

P(spam) = 150 /200 = 0.75 <br>

### 2. Probability of Not Spam

Number of Not Spam Emails:200−150=50 <br>
Formula:
P(Not Spam)=Number of Not Spam Emails/Total Email <br>

P(spam) = 50 / 200 = 0.25 <br>
