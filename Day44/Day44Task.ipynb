# **Task 4**
Research YOLOv3, YOLOv5 and YOLOv8.

1. YOLOv3

Released: 2018

Developed By: Joseph Redmon and collaborators

### Features
* Uses the Darknet-53 backbone network.
* Detects objects at three different scales.
* Improved accuracy over earlier YOLO versions.
* Suitable for real-time object detection.
### Advantages
* Faster than many traditional detectors.
* Good balance between speed and accuracy.
* Supports detection of small objects.
### Limitations
* Lower accuracy compared to newer versions.
* Larger model size.

2. YOLOv5

Released: 2020

Developed By: Ultralytics

### Features
* Implemented in PyTorch.
* Easy to train and deploy.
* Multiple model sizes (n, s, m, l, x).
* Faster training and inference.
### Advantages
* User-friendly.
* High speed and accuracy.
* Widely used in industry and research.
### Limitations
* Not an official release from the original YOLO authors.
* Newer versions provide better performance.

3. YOLOv8

Released: 2023

Developed By: Ultralytics

### Features
* Latest generation YOLO model.
* Anchor-free object detection.
### Supports:
* Object Detection
* Image Classification
* Instance Segmentation
* Pose Estimation
* Object Tracking
* Improved accuracy and speed.
### Advantages
* State-of-the-art performance.
* Easier training process.
* Better small-object detection.
* Flexible deployment options.
### Limitations
* Requires more computational resources than smaller models.
* Larger variants need powerful GPUs.

6. Retail and Inventory Management 
* Tracks products on shelves.

7. Robotics
