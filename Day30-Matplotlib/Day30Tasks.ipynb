```python
import matplotlib.pyplot as plt
```

**Task 2: Line Plot Practice** 
1. Create simple line graph  
2. Add labels  
3. Add title 

```python
#create simple line graph
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x,y)

#Add labels
plt.xlabel("X values")
plt.ylabel("Y values")
#Add title
plt.title("Simple Line Graph")
plt.show()
```

**Task 3: Styling Graphs** 
1. Add marker  
2. Change line style  
3. Add grid 

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
#Add marker and Change line style
plt.plot(x,y,marker='o', linestyle='--')

#Add labels
plt.xlabel("X values")
plt.ylabel("Y values")
#Add title
plt.title("Simple Line Graph")
#Add Grid
plt.grid()
plt.show()
```

**Task 4: Bar Chart Practice** 
1. Student marks comparison  
2. Sales comparison  
3. Product comparison  

```python
#Students marks comparison
students = ['Ann', 'Rahul', 'Ram', 'Arjun']
marks = [85, 90, 78, 88]

plt.bar(students, marks)

plt.title("Student Marks Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
#Sales Comparison
months = ['Jan', 'Feb', 'Mar', 'Apr','May','June']
sales = [5000, 7000, 6500, 8000,4000,9000]

plt.bar(months, sales)

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
#Product Comparison
product = ['Laptop','Iphone','Watch','Tablet']
price =[50000,100000,10000,20000]
plt.bar(product, price)

plt.title("Product Price Comparison")
plt.xlabel("Products")
plt.ylabel("Price")

plt.show()
```

**Task 5: Pie Chart Practice** 
1. Department distribution  
2. Expense analysis  
3. Course popularity  

```python
#Department distribution
departments = ['HR', 'IT', 'Sales', 'Marketing']
employees = [20, 40, 25, 15]

plt.pie(employees, labels=departments, autopct='%1.1f%%')

plt.title("Department Distribution")

plt.show()
```

```python
#Expense Analysis
expenses = ['Food', 'Rent', 'Travel', 'Shopping']
amount = [3000, 8000, 2000, 1500]

plt.pie(amount, labels=expenses, autopct='%1.1f%%')

plt.title("Monthly Expense Analysis")

plt.show()
```

```python
#Course Popularity
courses = ['Python', 'Data Analysis', 'AI', 'Data Science']
students = [45, 25, 55, 10]

plt.pie(students, labels=courses, autopct='%1.1f%%')

plt.title("Course Popularity")

plt.show()
```

**Task 6: Scatter Plot Practice**
1. Height vs weight  
2. Study hours vs marks  
3. Temperature vs sales 

```python
#Height vs weight
height = [150, 155, 160, 165, 170]
weight = [50, 55, 60, 65, 70]

plt.scatter(height, weight)

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")

plt.show()
```

```python
#Study hours vs marks
study_hours = [1, 2, 3, 4, 5]
marks = [50, 60, 70, 30, 40]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```
