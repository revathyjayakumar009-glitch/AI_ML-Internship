# 1. What is CSV file? 

A CSV (Comma-Separated Values) file is a simple file format used to store data in a table form.
Each line represents a row, and values are separated by commas.

**Example of CSV File**  

Name,Age,Marks<br>
Anu,20,85<br>
Rahul,21,90<br>
Meera,19,88<br>
**Explanation**  

* First row → Column names (headers)
* Next rows → Actual data
* Values are separated using commas (,)<br>
**Features of CSV Files**
  
* Easy to create and read
* Supported by Excel, Google Sheets, and Pandas
* Commonly used in Data Science and AI/ML projects
* Stores large datasets efficiently

# 3. What is read_csv()? 

**read_csv()** is a function used to read and load CSV files into Python.  
It converts the CSV data into a DataFrame, which makes data analysis easier.  

**Syntax**<br>
**pd.read_csv("filename.csv")**  

**Example**<br>
data = pd.read_csv("students.csv")

# 4. Difference between loc and iloc 

**Both loc[] and iloc[] are used to select data from a DataFrame**  
| Feature        | `loc[]`              | `iloc[]`             |
| -------------- | -------------------- | -------------------- |
| Selection Type | Label-based          | Index-based          |
| Uses           | Row/column names     | Row/column positions |
| Accepts        | Labels               | Integer indexes      |
| Example        | `data.loc[0,"Name"]` | `data.iloc[0,1]`     |
