# 1. Define Text Preprocessing. 

Text Preprocessing is the process of cleaning and transforming raw text into a structured and meaningful format so that computers can easily analyze and process it in Natural Language Processing (NLP) and Machine Learning applications.

#### It involves tasks such as:

* Converting text to lowercase
* Tokenization
* Removing stop words
* Removing punctuation and special characters
* Removing numbers
* Stemming and lemmatization (optional)

### Example:

Raw Text:<br>

Hello!!! I am learning NLP in 2026.<br>

After Text Preprocessing:<br>

hello learning nlp<br>

# 2. What is Tokenization?  

Tokenization is the process of breaking a piece of text into smaller units called tokens. These tokens can be words, sentences, or characters, depending on the NLP task.

It is one of the first and most important steps in Natural Language Processing (NLP) because it converts raw text into a format that computers can process.
### Example:

Original Text:<br>

Machine Learning is amazing.<br>

Word Tokenization:<br>

['Machine', 'Learning', 'is', 'amazing', '.']<br>

Sentence Tokenization:<br>

['Machine Learning is amazing.']<br>

### Types of Tokenization
* Word Tokenization – Splits text into individual words.
* Sentence Tokenization – Splits text into separate sentences.
* Character Tokenization – Splits text into individual characters.
### Importance of Tokenization
* Converts text into meaningful units.
* Makes text easier for computers to understand.
* Serves as the foundation for NLP tasks such as sentiment analysis, text classification, machine translation, and chatbots.

# 3. What are Stop Words? 

Stop Words are commonly used words in a language that usually do not add significant meaning to a sentence. In many Natural Language Processing (NLP) tasks, these words are removed to focus on the important words in the text.
### Examples of Stop Words
* is
* am
* are
* was
* were
* the
* a
* an
* of
* in
* on
* at
* to
* for
### Example

Original Sentence:<br>

The cat is sitting on the mat.<br>

After Removing Stop Words:<br>

cat sitting mat<br>

# 4. Why remove punctuation? 

Removing punctuation is a text preprocessing step in Natural Language Processing (NLP) that eliminates symbols such as . , ! ? : ; ( ) " ' - because they often do not contribute to the meaning of the text.

### Reasons for Removing Punctuation
* Removes unnecessary symbols (noise) from the text.
* Makes text easier for computers to process.
* Improves the accuracy of NLP and Machine Learning models.
* Helps in consistent text analysis and tokenization.
* Reduces the size and complexity of the text data.

### Example

Original Text:<br>

Hello! Welcome to AI, Machine Learning.<br>

After Removing Punctuation:<br>

Hello Welcome to AI Machine Learning<br>

# 5. Why convert text to lowercase? 

Converting text to lowercase is a common text preprocessing step in Natural Language Processing (NLP). It changes all uppercase letters to lowercase so that words with different letter cases are treated as the same word.

### Reasons for Converting Text to Lowercase
* Ensures consistency in the text.
* Treats words like "AI", "Ai", and "ai" as the same word.
* Reduces duplicate words caused by different letter cases.
* Improves the accuracy of NLP and Machine Learning models.
* Simplifies text analysis and processing.

### Example

Original Text:<br>

Python is FUN. PYTHON is Easy.<br>

After Converting to Lowercase:<br>

python is fun. python is easy.<br>
