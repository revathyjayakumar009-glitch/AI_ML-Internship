# 1. Define NLP. 

Natural Language Processing (NLP) is a branch of Artificial Intelligence (AI) that enables computers to understand, interpret, process, and generate human language (such as text and speech) in a meaningful way.

Example: Chatbots like ChatGPT, Siri, and Alexa use NLP to understand user queries and provide appropriate responses.

# 2. Explain Intent Recognition.  

Intent Recognition is the process of identifying the purpose or goal behind a user's message. It helps a chatbot understand what the user wants so it can provide the correct response.

### Example:

* User: "Book a flight to Delhi."
* Intent: Book Flight

### Another example:

* User: "Track my order."
* Intent: Track Order

### Importance:

* Understands the user's request.
* Selects the appropriate response.
* Improves chatbot accuracy and user experience.

# 3. Explain Entity Recognition.  

Entity Recognition (also called Named Entity Recognition – NER) is the process of identifying and extracting important information (entities) from a user's message. These entities provide the specific details needed to fulfill the user's request.

### Example:

* User: "Book a flight to Mumbai tomorrow."

* Intent: Book Flight
* Entities:
Location: Mumbai<br>
Date: Tomorrow

### Another example:

* User: "Transfer ₹5000 to Rahul."

* Intent: Money Transfer
* Entities:
Amount: ₹5000<br>
Person: Rahul

### Importance:

* Extracts key information from user input.
* Helps chatbots perform tasks accurately.
* Improves the quality and relevance of responses.

# 4. Differentiate Stemming and Lemmatization. 

| **Stemming**                                                                    | **Lemmatization**                                                        |
| ------------------------------------------------------------------------------- | ------------------------------------------------------------------------ |
| Reduces a word to its root by removing prefixes or suffixes using simple rules. | Converts a word to its dictionary (base) form using linguistic analysis. |
| May produce non-dictionary words.                                               | Produces valid dictionary words.                                         |
| Faster and simpler.                                                             | More accurate but slower.                                                |
| Does not consider the word's context or meaning.                                | Considers the word's context and meaning.                                |


# 5. Explain the NLP workflow in chatbots.  
The NLP workflow in chatbots is the sequence of steps a chatbot follows to understand a user's message and generate an appropriate response.

```python
User Input
      ↓
Text Preprocessing
      ↓
Tokenization
      ↓
Stop Word Removal
      ↓
Intent Recognition
      ↓
Entity Recognition
      ↓
Generate Response
      ↓
Reply to User
```
