# 1. What is Decision tree?
A Decision Tree is a supervised machine learning algorithm used for both classification and regression tasks. It predicts an output by learning a series of decision rules from the training data.<br>

The algorithm represents decisions in the form of a tree-like structure, where:<br>

* Root Node → Starting point of the tree
* Decision Nodes → Conditions used to split the data
* Branches → Outcomes of decisions
* Leaf Nodes → Final prediction or result<br>
**How It Works**<br>

A Decision Tree repeatedly asks questions about the input features and splits the data into smaller groups. This process continues until the data is classified into the correct category or a prediction is made.<br>
#### Real-Life Example

When deciding whether to carry an umbrella:<br>
Is it cloudy?<br>
      /     \ <br>
    Yes      No<br>
     |        |<br>
Will it rain?  No Umbrella<br>
     /   \ <br>
   Yes    No<br>
    |      |<br>
Umbrella  No Umbrella<br>

# 2. What is a Root Node? 
A Root Node is the topmost node of a Decision Tree and serves as the starting point for all decision-making processes. It contains the entire dataset before any splitting occurs.<br>

The root node represents the first and most important decision because it splits the data into groups based on the feature that best separates the classes.<br>
Example:<br>
Study Hours > 3?<br>
      /      \ <br>
    Yes       No<br>

    here Student > 3 ? is the root node

# 3. What is a Branch?  
A Branch is a connection between nodes in a Decision Tree that represents the outcome of a decision or condition. It shows the path followed from one node to another based on whether a condition is true or false.<br>

Branches help divide the dataset into smaller subsets and guide the tree toward a final prediction.<br>
Example:<br>
Study Hours > 3?<br>
      /      \ <br>
    Yes       No<br>
In this example:<br>

* Yes is a branch.
* No is a branch.<br>

These branches connect the root node to the leaf nodes.

# 4. What is a Leaf Node? 
A Leaf Node is the final node in a Decision Tree that provides the prediction or outcome. It is also called a Terminal Node because no further splitting or decisions occur after reaching it.<br>

A leaf node represents the final class in a classification problem or the final value in a regression problem.<br>
Example<br>
Study Hours > 3?<br>
      /      \ <br>
    Yes       No <br>
     |         | <br>
   Pass      Fail <br>
In this example: <br>

* Pass → Leaf Node <br>
* Fail → Leaf Node <br>

These nodes provide the final prediction and do not split further. <br> 

# 5. Advantages of Decision Trees?  
Decision Trees are widely used in machine learning because they are simple, powerful, and easy to interpret. They can be applied to both classification and regression problems.<br>

**1. Easy to Understand and Interpret** <br>

Decision Trees follow a tree-like structure that resembles human decision-making. <br>
**2. Simple to Visualize** <br>

The entire model can be displayed as a tree diagram, making it easy to analyze the decision-making process.<br>

This visualization helps identify:<br>

* Important features
* Decision paths
* Final outcomes

**3. Requires Little Data Preprocessing** <br>

Decision Trees do not require:<br>

❌ Feature Scaling <br>
❌ Normalization <br>
❌ Standardization <br>

They can work directly with raw data.<br>

**4. Handles Both Numerical and Categorical Data**<br>

Decision Trees can process:<br>

**Numerical Data** <br>
* Age
* Salary
* Marks
**Categorical Data**
* Gender
* City
* Product Category<br>


This flexibility makes them suitable for many real-world applications.<br>
**5. Works for Both Classification and Regression**<br>
**Classification**

Predict categories such as:<br>

* Pass/Fail
* Spam/Not Spam
* Disease/No Disease<br>
**Regression**<br>

Predict continuous values such as:<br>

* House Prices
* Sales Revenue
* Salary<br>
**6.Identifies Important Features Automatically**

The algorithm automatically selects the most informative features for splitting.<br>
**7. Handles Non-Linear Relationships**

* Many real-world datasets have complex relationships between variables.

* Decision Trees can capture these non-linear patterns without requiring mathematical transformations.<br>
**8. Fast Prediction**

* Once the tree is trained, making predictions is very fast because the model only follows a path from the root node to a leaf node.

This makes Decision Trees useful for:<br>

* Real-time systems
* Recommendation systems
* Fraud detection<br>
**9. Works with Missing Values (in Some Implementations)** <br>

* Certain Decision Tree implementations can handle missing data effectively, reducing the need for extensive preprocessing.<br>
**10. Useful for Feature Selection** <br>

Since the tree chooses the most important features first, it provides insights into which variables have the greatest impact on predictions.<br>
