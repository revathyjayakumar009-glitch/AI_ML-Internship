**Task 4: Random Choice** 
1. Pick random color  
2. Generate multiple choices  
3. Simulate random names 

```python
#Pick random color
colors = ["Red", "Blue", "Green", "Yellow"]
print(np.random.choice(colors))
```

```python
#Generate multiple choices
numbers = [1, 2, 3, 4, 5]

choices = np.random.choice(numbers, size=3)

print(choices)
```

```python
#simulate random names
names = ["Arun", "Meera", "Rahul", "Anu", "Kiran"]

random_names = np.random.choice(names, size=3)

print(random_names)
```

**Task 5: Shuffle** 
1. Shuffle array  
2. Shuffle student IDs 

```python
#Shuffle array
arr = np.array([1, 2, 3, 4, 5])
np.random.shuffle(arr)
print(arr)
```

```python
#Shuffle student IDs
student_ids = np.array([101, 102, 103, 104, 105])
np.random.shuffle(student_ids)
print(student_ids)
```
