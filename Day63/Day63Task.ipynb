# Practical Task 4 
Write a Python program that predicts an intent using a simple dictionary lookup.

```python
# Dictionary containing keywords and corresponding intents
intent_dict = {
    "hello": "Greeting",
    "hi": "Greeting",
    "fees": "Fees Inquiry",
    "fee": "Fees Inquiry",
    "duration": "Course Duration",
    "course": "Course Duration",
    "eligibility": "Eligibility",
    "eligible": "Eligibility",
    "location": "College Location",
    "address": "College Location",
    "contact": "Contact Information",
    "phone": "Contact Information",
    "admission": "Admission Inquiry",
    "hostel": "Hostel Inquiry",
    "placement": "Placement Inquiry",
    "thanks": "Thanks",
    "thank": "Thanks",
    "bye": "Goodbye"
}

# Get user input
message = input("Enter your message: ").lower()

# Predict intent using dictionary lookup
intent = "Unknown"

for keyword, label in intent_dict.items():
    if keyword in message:
        intent = label
        break

# Display result
print("Predicted Intent:", intent)
```
