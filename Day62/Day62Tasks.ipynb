Write a Python program to tokenize the sentence:<br> 
Artificial Intelligence is changing the world.

```python


sentence = "Artificial Intelligence is changing the world."

tokens = word_tokenize(sentence)

print("Tokens:", tokens)
```

 # **Practical Task 3** 
Remove stop words from: <br>
I want to book a train ticket to Delhi.

```python
text = "I want to book a train ticket to Delhi" 
tokens = word_tokenize(text) 
stop_words = set(stopwords.words("english")) 
filtered = [] 
for word in tokens:
    if word.lower() not in stop_words: 
        filtered.append(word)
        print(filtered)
```

 # **Practical Task 4** 
Apply stemming to: <br>
• Playing

• Reading

```python
stemmer = PorterStemmer() 
print(stemmer.stem("Playing")) 
print(stemmer.stem("Running"))
print(stemmer.stem("Reading")) 
print(stemmer.stem("Learning")) 
```

# **Practical Task 5** 
Build a chatbot that answers questions about:<br> 
• Fees  
• Course Duration  
• Eligibility  
• Contact Number  
• College Location<br>  
using NLTK tokenization.

```python
print("===== College Enquiry Chatbot =====")
print("Type 'exit' to quit.\n")

while True:
    user = input("You: ").lower()

    if user == "exit":
        print("Chatbot: Thank you! Have a great day.")
        break

    # Tokenize user input
    tokens = word_tokenize(user)

    if "fees" in tokens or "fee" in tokens:
        print("Chatbot: The course fee is ₹50,000 per year.")

    elif "duration" in tokens:
        print("Chatbot: The course duration is 3 years.")

    elif "eligibility" in tokens or "eligible" in tokens:
        print("Chatbot: Candidates must have passed 12th grade with the required eligibility criteria.")

    elif "contact" in tokens or "number" in tokens or "phone" in tokens:
        print("Chatbot: You can contact us at +91-9876543210.")

    elif "location" in tokens or "college" in tokens or "address" in tokens:
        print("Chatbot: Our college is located in Ernakulam, Kerala.")

    else:
        print("Chatbot: Sorry, I didn't understand your question.")
```
