**Basic Aggregation** 
1. Find sum  
2. Find mean  
3. Find max and min 

```python
import numpy as np
```

```python
arr=np.array([5,4,8,7,10,6,2])
```

```python
print("Sum:",np.sum(arr))  #sum
print("Mean:",np.mean(arr)) #mean
print("Max:",np.max(arr),"Min:",np.min(arr))  #max and min
```

**Task 3: Statistical Functions** 
1. Find median  
2. Find standard deviation  
3. Find variance 

```python
print("Median:",np.median(arr))  #median
print("Standard deviation:",round(np.std(arr),2)) #standard deviation
print("Variance:",np.var(arr))  #varience
```

**Task 4: Axis Practice** 
1. Row-wise sum  
2. Column-wise sum  
3. Mean using axis 

```python
arr1 = np.array([[2,4,6],
               [8,10,12]])
print(arr1)
```

```python
print("Row-wise sum:",np.sum(arr1,axis=1)) #row wise sum
print("Column-wise sum:",np.sum(arr1,axis=0)) #Column wise sum
print("Mean using axis:",np.mean(arr1,axis=1))
print("Mean using axis:",np.mean(arr1,axis=0))
```

**Task 5: Index Functions** 
1. Find index of max  
2. Find index of min 

```python
arr=np.array([5,4,8,7,10,6,2])
print("Index of max:",np.argmax(arr)) #Index of max
print("Index of min:",np.argmin(arr)) #Index of min
```

**Task 6: Cumulative** 
1. Compute cumulative sum  
2. Analyze output

```python
arr=np.array([5,4,8,7,10,6,2])
print(np.cumsum(arr))  #cumilative sum
```
