# 1. What is Computer Vision? 

Computer Vision is a branch of Artificial Intelligence (AI) that enables computers to interpret, analyze, and understand visual information from images and videos, similar to how humans use their eyes and brain.

#### Example:

* Detecting faces in photos
* Recognizing vehicles in traffic cameras
* Identifying diseases from medical images
* Detecting objects in self-driving cars

# 2. What is Object Detection?  

Object Detection is a Computer Vision technique that identifies and locates objects within an image or video.

It not only recognizes what the object is but also where it is located by drawing a bounding box around it.

#### Example

In a street image, an object detection model can detect:

* Cars 
* People 
* Traffic Lights 
* Bicycles 

and place a bounding box around each detected object.

#### Key Tasks of Object Detection
* Classification – Identifying the object's category (e.g., car, person, dog).
* Localization – Determining the object's position using bounding boxes.
#### Applications
* Self-driving vehicles
* Security and surveillance
* Face detection
* Medical image analysis
* Robotics

# 3. What is YOLO? 

YOLO (You Only Look Once) is a popular real-time object detection algorithm used in Computer Vision to identify and locate multiple objects in an image or video quickly and accurately.

Unlike traditional methods that examine an image multiple times, YOLO processes the entire image in a single pass, making it extremely fast and suitable for real-time applications.

### Key Features of YOLO
* Fast and efficient object detection
* Detects multiple objects simultaneously
* High accuracy
* Works in real-time on images and videos
* Uses a single neural network for detection
### How YOLO Works
* Divides the image into a grid.
* Predicts bounding boxes for objects.
* Assigns confidence scores and class labels.
* Outputs the detected objects with their locations.
### Applications
* Self-driving cars
* Security surveillance
* Traffic monitoring
* Robotics
* Healthcare imaging

# 4. What is a Bounding Box?  

A Bounding Box is a rectangular box drawn around an object in an image or video to indicate its location.

In object detection, bounding boxes help identify where an object is present within the image.

### Example

If YOLO detects a car in an image, it draws a rectangle around the car and labels it as "Car".

### Bounding Box Information

A bounding box is usually represented by:

* x-coordinate (horizontal position)
* y-coordinate (vertical position)
* Width of the box
* Height of the box
### Purpose of Bounding Boxes
* Locate objects accurately
* Separate multiple objects in an image
* Help in object tracking and recognition
### Applications
* Vehicle detection
* Face detection
* Pedestrian tracking
* Medical image analysis

# 5. Advantages of YOLO? 

YOLO (You Only Look Once) is one of the most widely used object detection algorithms because of its speed and efficiency.

### Advantages
1. Real-Time Detection

YOLO can detect objects instantly, making it ideal for live video streams and real-time applications.

2. High Speed

Since YOLO processes the entire image in a single pass, it is much faster than traditional object detection methods.

3. Detects Multiple Objects

YOLO can identify and locate multiple objects in the same image simultaneously.

4. Good Accuracy

It provides accurate object detection while maintaining high processing speed.

5. End-to-End Learning

YOLO uses a single neural network to perform both object classification and localization.

6. Global Image Understanding

YOLO analyzes the whole image at once, reducing false detections caused by background regions.

7. Suitable for Various Applications

It is widely used in:

* Self-driving cars
* Security surveillance
* Robotics
* Traffic monitoring
* Healthcare imaging
