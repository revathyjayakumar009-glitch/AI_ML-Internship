```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

**Task 2: Line Plot Practice** 
1. Create line graph  
2. Add style  
3. Compare data trends  

```python
# Create linr graph
x = [1, 2, 3, 4, 5]
y = [10, 15, 20, 18, 25]

sns.lineplot(x=x, y=y)

plt.title("Simple Line Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
#Add Style
sns.set_style("darkgrid")

x = [1, 2, 3, 4, 5]
y = [10, 15, 20, 18, 25]

sns.lineplot(x=x, y=y, marker="o")

plt.title("Styled Line Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
#Compare data trends
months = [1, 2, 3, 4, 5]

sales_2024 = [20, 25, 30, 35, 40]
sales_2025 = [18, 28, 32, 38, 45]

sns.lineplot(x=months, y=sales_2024, marker="o", label="2024")
sns.lineplot(x=months, y=sales_2025, marker="o", label="2025")

plt.title("Sales Trend Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()

plt.show()
```

**Task 3: Bar Plot Practice** 
1. Student marks comparison  
2. Sales comparison  
3. Employee salary comparison 

```python
#Student mark comparison
students = ["Arun", "Rahul", "Sneha", "Anu"]
marks = [85, 92, 78, 88]

sns.barplot(x=students, y=marks)

plt.title("Student Marks Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
#Sales Comparison
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [50, 80, 35, 45]

sns.barplot(x=products, y=sales)

plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
#Employee Salary Comparison
employees = ["John", "David", "Sara", "Maya"]
salary = [40000, 55000, 48000, 62000]

sns.barplot(x=employees, y=salary)

plt.title("Employee Salary Comparison")
plt.xlabel("Employees")
plt.ylabel("Salary")

plt.show()
```

**Task 4: Scatter Plot Practice**
1. Height vs weight  
2. Study hours vs marks

```python
#Height vs weight
height = [150, 155, 160, 165, 170, 175]
weight = [45, 50, 55, 60, 68, 72]

sns.scatterplot(x=height, y=weight)

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")

plt.show()
```

```python
# Temparature vs Sales
temperature = [20, 22, 25, 28, 30, 35]
sales = [100, 120, 150, 180, 90, 50]

sns.scatterplot(x=temperature, y=sales)

plt.title("Temperature vs Sales")
plt.xlabel("Temperature")
plt.ylabel("Sales")

plt.show()
```

**Task 5: Histogram Practice** 
1. Create histogram  
2. Analyze frequency  
3. Use random dataset 

```python
#Create histogram
data = [10, 12, 15, 18, 20, 22, 25, 25, 30, 35]

sns.histplot(data)

plt.title("Simple Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
#Analyze frequency
marks = [40, 45, 50, 55, 60, 65, 70, 70, 75, 80, 85, 90]

sns.histplot(marks, bins=5)

plt.title("Marks Frequency Analysis")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
#Use random dataset
random_data = np.random.randint(1, 100, 100)

sns.histplot(random_data, bins=10)

plt.title("Random Dataset Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

**Task 6: Box Plot Practice** 
1. Detect outliers  
2. Analyze salary dataset  
3. Analyze marks dataset  

```python
#Detect outlier
data = [10, 12, 14, 15, 18, 20, 22, 25, 100]

sns.boxplot(x=data)

plt.title("Outlier Detection")

plt.show()
```

The value 100 appears as an outlier

```python
#Analyze salary dataset
salary = [25000, 30000, 32000, 35000, 40000, 45000, 50000, 120000]

sns.boxplot(x=salary)

plt.title("Salary Dataset Analysis")
plt.xlabel("Salary")

plt.show()
```

120000 is an outlier compared to other salaries.

```python
#Analyze mark dataset
marks = [45, 50, 55, 60, 65, 70, 75, 80, 95]

sns.boxplot(x=marks)

plt.title("Marks Dataset Analysis")
plt.xlabel("Marks")

plt.show()
```

Helps understand spread, median, and distribution.

**Task 7: Count Plot Practice**
1. Course count  
2. Department count  
3. Product category count

```python
#Course count
courses = ["Python", "Java", "Python", "AI", "Java", "Python", "AI"]

sns.countplot(x=courses)

plt.title("Course Count")
plt.xlabel("Courses")
plt.ylabel("Count")

plt.show()
```

```python
#Department Count
departments = ["HR", "IT", "Sales", "IT", "HR", "Marketing", "IT"]

sns.countplot(x=departments)

plt.title("Department Count")
plt.xlabel("Departments")
plt.ylabel("Count")

plt.show()
```

```python
#Product Category Count
categories = ["Electronics", "Clothing", "Electronics", "Food",
              "Clothing", "Electronics", "Food"]

sns.countplot(x=categories)

plt.title("Product Category Count")
plt.xlabel("Categories")
plt.ylabel("Count")

plt.show()
```
