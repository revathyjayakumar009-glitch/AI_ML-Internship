# 1. What are Pre-trained Word Embeddings? 

Pre-trained Word Embeddings are word vectors that have already been trained on large text datasets (such as Wikipedia or Common Crawl). Instead of training embeddings from scratch, these pre-trained vectors can be directly used in NLP applications to represent words based on their meanings and relationships.

### Key benefits:

* Saves training time
* Improves model accuracy
* Captures semantic relationships between words
* Works well even with small datasets

# 2. Explain GloVe.  

GloVe (Global Vectors for Word Representation) is a pre-trained word embedding model developed by Stanford University. It learns word vector representations by analyzing the global co-occurrence statistics of words in a large text corpus.

Words that appear in similar contexts are assigned similar vector representations, helping NLP models understand semantic relationships.

### Advantages:

Captures global context of words
Produces high-quality word embeddings
Efficient and widely used in NLP tasks

### Limitation:

Cannot generate embeddings for out-of-vocabulary (OOV) words that were not seen during training.

# 3. Explain FastText.  

FastText is a pre-trained word embedding model developed by Meta AI (Facebook AI Research). Unlike GloVe and Word2Vec, FastText represents each word as a collection of character n-grams (subwords) instead of treating it as a single unit.

This allows FastText to generate embeddings for unseen or out-of-vocabulary (OOV) words by using their character-level information.

### Advantages:

* Handles unseen (OOV) words effectively
* Supports morphologically rich languages
* Works well with spelling variations and rare words
* Produces high-quality word embeddings for many NLP tasks

# 4. What is OOV?  

OOV (Out-of-Vocabulary) refers to words that were not present in the training vocabulary of a word embedding model.

Since the model has never seen these words during training, it cannot generate embeddings for them (in models like GloVe). However, FastText can handle OOV words by creating embeddings from their character n-grams.

### Example:

* Training vocabulary: cat, dog, apple
* OOV word: chatbot

Here, chatbot is an OOV word because it was not included in the training data.

| Training method    | CBOW & Skip-Gram  | Global word co-occurrence        | Character n-grams (subwords)                                        |
| Uses context       | Local context     | Global context                   | Local context + subword information                                 |
| Handles OOV words  | No                | No                               | Yes                                                                 |
| Morphology support | Limited           | Limited                          | Excellent                                                           |
| Best for           | General NLP tasks | Capturing semantic relationships | Rare words, spelling variations, and morphologically rich languages |
