# 1. What is a class? 

A class is the category or label assigned to a detected object in an image.

### Examples of classes:

* Person
* Car
* Dog
* Cat
* Bicycle
* Horse

When YOLO detects an object, it predicts which class the object belongs to.

# 2. What is a bounding box? 

A bounding box is a rectangular box drawn around a detected object in an image to show its location.

### Example:

* A box around a person indicates where the person is in the image.
* A box around a car indicates the car's position.

### Purpose:

* Identifies the exact location of an object.
* Helps in object detection and tracking.

# 3. What is confidence score? 

A confidence score is the probability or percentage that indicates how certain the YOLO model is about its prediction.
### Example:
Person → 98% <br>
Dog → 95% <br>
Car → 90% <br>
Here, 98%, 95%, and 90% are confidence scores. <br>
### Interpretation:

* 90–100% → Very high confidence
* 70–89% → High confidence
* 50–69% → Moderate confidence
* Below 50% → Low confidence

# 4. What is confidence threshold? 

A confidence threshold is the minimum confidence score required for a detected object to be accepted and displayed by the YOLO model.
### Example:

* Confidence Threshold = 50%
* Person → 95% ✅ (Displayed)
* Dog → 82% ✅ (Displayed)
* Cat → 45% ❌ (Ignored)

Since the cat's confidence score is below the threshold, it is not shown in the final results.

# 5. Why are bounding boxes important? 


Bounding boxes are important because they show the exact location and size of detected objects in an image or video.

Importance of Bounding Boxes:<br>
✅ Locate Objects – Identify where an object is present in the image.<br>
✅ Object Tracking – Help track moving objects across video frames.<br>
✅ Measure Size and Position – Provide information about an object's dimensions and coordinates.<br>
✅ Enable Real-World Applications – Used in autonomous vehicles, surveillance systems, robotics, and medical imaging.<br>
✅ Improve Analysis – Allow models to distinguish between multiple objects in the same image.<br>
