# 1. What is Logistic Regression? 

Logistic Regression is a supervised machine learning algorithm used for classification problems, especially when the output has two possible categories (binary classification).

It predicts the probability that a data point belongs to a particular class and then assigns it to a category based on that probability.<br>

**Examples:**<br>
* Email → Spam or Not Spam<br>
* Loan → Approved or Rejected<br>
* Disease → Positive or Negative<br>
* Customer → Churn or Not Churn<br>
**Key Points:** <br>
* Used for classification, not prediction of continuous values.<br>
* Produces output between 0 and 1 (probability).<br>
* Uses the Sigmoid (Logistic) Function to convert predictions into probabilities.<br>
* One of the most widely used algorithms for binary classification tasks.<br>

# 2. Why is it used? 

Reasons for Using Logistic Regression:<br>

✅ Classification Tasks<br>

* Determines which class an input belongs to.<br>
* Example: Spam or Not Spam.<br>

✅ Probability Prediction<br>

* Provides the probability of an event occurring.<br>
* Example: A customer has an 85% chance of buying a product.<br>

✅ Simple and Efficient<br>

* Easy to understand, implement, and train.<br>

✅ Works Well for Binary Outcomes<br>

* Ideal for Yes/No, True/False, Pass/Fail problems.<br>

✅ Interpretable Results<br>

* Helps understand how each feature affects the prediction.<br>

# 3. What is Binary Classification? 

Binary Classification is a type of machine learning problem where the model predicts one of two possible classes or categories.<br>

The output can only have two outcomes.(0 and 1)<br>
**Examples:** <br>
| Problem           | Class 1  | Class 2   |
| ----------------- | -------- | --------- |
| Email Filtering   | Spam     | Not Spam  |
| Exam Result       | Pass     | Fail      |
| Disease Detection | Positive | Negative  |
| Loan Approval     | Approved | Rejected  |
| Fraud Detection   | Fraud    | Not Fraud |


# 4. What is a Class Label? 

A Class Label is the target output (category) that a machine learning model predicts in a classification problem.


# 5. Difference between Linear and Logistic Regression? 

| Feature                | Linear Regression                         | Logistic Regression                   |
| ---------------------- | ----------------------------------------- | ------------------------------------- |
| **Purpose**            | Predicts continuous values                | Predicts categories/classes           |
| **Problem Type**       | Regression                                | Classification                        |
| **Output**             | Numeric value                             | Probability (0 to 1) and class label  |
| **Output Range**       | Any value (-∞ to +∞)                      | Between 0 and 1                       |
| **Function Used**      | Linear Equation                           | Sigmoid Function                      |
| **Target Variable**    | Continuous                                | Categorical                           |
| **Examples**           | Salary Prediction, House Price Prediction | Spam Detection, Disease Prediction    |
| **Evaluation Metrics** | MAE, MSE, RMSE, R² Score                  | Accuracy, Precision, Recall, F1-Score |
