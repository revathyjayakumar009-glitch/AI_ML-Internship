# 1. What is missing data?  

Missing data means some information or values are not available in a dataset.  
In Python Pandas, missing values are usually represented as:<br>
Nan --> Not a number
**Why Missing Data Happens?**
* Data not entered
* Survey forms left blank
* System errors
* Data lost during collection
* Importance in AI/ML

**Handling missing data is important because:**

* It improves data quality
* Helps build accurate ML models
* Prevents errors during analysis

# 2. Why data cleaning is important? 

Data cleaning is the process of fixing or removing incorrect, missing, duplicate, or messy data from a dataset.<br>

Clean data is very important in AI/ML and Data Analysis because machine learning models depend on good-quality data.

#### Importance of Data Cleaning
**1. Improves Accuracy**

Clean data gives better and more accurate results.<br>

**2. Removes Errors**

It helps remove:

* Missing values
* Duplicate rows
* Wrong data entries
* Extra spaces and formatting issues<br>
**3.Better Machine Learning Models**

AI/ML models learn from data.<br>
If the data is bad, the model performance will also be poor.<br>

**4. Saves Time**

Clean datasets are easier to analyze and visualize.<br>

**5. Helps in Decision Making**

Organizations use clean data to make correct business decisions.

# 3. What is isnull()? 

isnull() is a Pandas function used to detect missing values in a dataset.<br>

It checks each value and returns:

* True → if the value is missing (NaN)
* False → if the value is present

# 4. What is dropna()? 

dropna() is a Pandas function used to remove missing values (NaN) from a dataset.

It can:

* Remove rows with missing values
* Remove columns with missing values

# 5. What is fillna()? 

fillna() is a Pandas function used to replace missing values (NaN) with another value.

Instead of removing data, it fills the empty places with:

A fixed value
* Mean
* Median
* Custom text, etc.
