# **Practical Task 2**
Stem the following words: <br>
Playing <br>
Working <br>
Learning <br>
Running<br>

```python
from nltk.stem import PorterStemmer
```

```python
# Create a stemmer object
stemmer = PorterStemmer()

# List of words
words = ["Playing", "Working", "Learning", "Running"]

# Apply stemming
for word in words:
    print(word, "->", stemmer.stem(word))
```

# **Practical Task 3**
Lemmatize: <br>
Cars <br>
Children <br>
Mice <br>
Dogs<br>

```python
from nltk.stem import WordNetLemmatizer
import nltk
```

```python
nltk.download('wordnet')
```

```python
# Create a lemmatizer object
lemmatizer = WordNetLemmatizer()

# List of words
words = ["Cars", "Children", "Mice", "Dogs"]

# Apply lemmatization
for word in words:
    print(word, "->", lemmatizer.lemmatize(word.lower()))
```

# **Practical Task 4**
Compare outputs of: <br>
• Stemming  <br>
• Lemmatization <br> 
for the same words.<br>

```python
# Compare outputs
print(f"{'Word':<12}{'Stemming':<15}{'Lemmatization'}")
print("-" * 40)

for word in words:
    print(f"{word:<12}{stemmer.stem(word):<15}{lemmatizer.lemmatize(word)}")
```
