# 1. What is Random Forest? 

Random Forest is a supervised machine learning algorithm that builds multiple Decision Trees and combines their predictions to make a final decision.

For classification, it uses majority voting:

* Tree 1 → Pass
* Tree 2 → Pass
* Tree 3 → Fail
* Final Prediction → Pass

For regression, it uses the average of predictions from all trees.

#### Key Features
* Uses many Decision Trees instead of one.
* Improves accuracy and reliability.
* Reduces overfitting compared to a single Decision Tree.
* Works well for both classification and regression problems.
#### Example

A bank can use Random Forest to predict whether a loan application should be Approved or Rejected based on factors like income, credit score, and employment history.

# 2. What is an Ensemble Method? 

An Ensemble Method is a machine learning technique that combines the predictions of multiple models to produce a better and more accurate result than a single model.

#### Why Use Ensemble Methods?
* Improves prediction accuracy
* Reduces overfitting
* Increases model stability and reliability
#### Example

Suppose three models predict a student's result:

* Model 1 → Pass
* Model 2 → Pass
* Model 3 → Fail

Using majority voting, the final prediction is Pass.

#### Random Forest as an Ensemble Method

Random Forest is an ensemble method because it combines the predictions of many Decision Trees and uses majority voting (for classification) or averaging (for regression) to make the final prediction.

# 3. What is Voting?  

Voting is a technique used in ensemble learning where multiple models make predictions, and the final prediction is chosen based on the majority vote.

#### How Voting Works

Suppose five Decision Trees make predictions:
| Tree   | Prediction |
| ------ | ---------- |
| Tree 1 | Pass       |
| Tree 2 | Pass       |
| Tree 3 | Fail       |
| Tree 4 | Pass       |
| Tree 5 | Fail       |
* Pass = 3 votes
* Fail = 2 votes

Final Prediction = Pass (majority vote)

#### Why Voting is Used
* Improves prediction accuracy
* Reduces errors from individual models
* Makes the model more reliable
### In Random Forest

Each Decision Tree gives a prediction, and the class with the most votes becomes the final output. This process is called majority voting.

# 5. Difference between Decision Tree and Random Forest? 

| Decision Tree                          | Random Forest                                                                         |
| -------------------------------------- | ------------------------------------------------------------------------------------- |
| Uses a **single tree** for prediction. | Uses **multiple decision trees** for prediction.                                      |
| More prone to **overfitting**.         | Reduces overfitting by combining many trees.                                          |
| Less accurate on complex datasets.     | Generally more accurate and reliable.                                                 |
| Faster to train and predict.           | Slightly slower due to multiple trees.                                                |
| Easy to understand and visualize.      | Harder to interpret because it contains many trees.                                   |
| Prediction comes from one tree.        | Prediction comes from **majority voting** (classification) or averaging (regression). |
