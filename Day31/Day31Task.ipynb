```python
import matplotlib.pyplot as plt
import numpy as np
```

**Task 2: Multiple Graphs** 
1. Plot two lines  
2. Add labels  
3. Add legend  

```python
x  = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [40, 30, 20, 10]
#Plot two lines
plt.plot(x, y1, label="Sales")
plt.plot(x, y2, label="Profit")

#Add labels
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Value")
#Add legend
plt.legend()

plt.show()
```

```python
**Task 3: Figure Customization** 
1. Change figure size  
2. Add title  
3. Add grid 
```

```python
x  = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [40, 30, 20, 10]
#Change figure size
plt.figure(figsize=(8, 5))
plt.plot(x, y1, label="Sales",  color="blue",   linestyle="solid",  marker="o", linewidth=2)
plt.plot(x, y2, label="Profit", color="orange", linestyle="dashed", marker="o", linewidth=2)
#Add Title
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Value")
plt.legend()
#Add grid
plt.grid(True)
plt.show()
```

**Task 4: Subplot Practice**
1. Create 2 subplots  
2. Create 4 subplots  
3. Compare graphs  

```python
#create 2 subplots
x = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [5, 15, 25, 35]

fig ,ax=plt.subplots(1, 2, figsize=(10,4))   #fig ---> all note book page   ax--> one graph/ chart on the page

ax[0].plot(x, y1)
ax[0].set_title("Sales Graph")

ax[1].plot(x, y2)
ax[1].set_title("Profit Graph")

plt.tight_layout()
plt.show()
```

```python
#create 4 subplots
fig, ax = plt.subplots(2, 2, figsize=(10,8))

# Plot 1
ax[0,0].plot([1,2,3], [2,4,6])
ax[0,0].set_title("Line Plot")

# Plot 2
ax[0,1].bar([1,2,3], [3,5,7])
ax[0,1].set_title("Bar Plot")

# Plot 3
ax[1,0].scatter([1,2,3], [7,5,3])
ax[1,0].set_title("Scatter Plot")

# Plot 4
ax[1,1].pie([30,40,30],
            labels=['A','B','C'],
            autopct='%1.1f%%')
ax[1,1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
#Compare graphs
months = ['Jan', 'Feb', 'Mar', 'Apr']

sales = [200, 250, 300, 280]
profit = [50, 70, 90, 60]

fig, ax = plt.subplots(1, 2, figsize=(10,4))

# Sales graph
ax[0].plot(months, sales, marker='o')
ax[0].set_title("Monthly Sales")

# Profit graph
ax[1].bar(months, profit)
ax[1].set_title("Monthly Profit")

plt.tight_layout()
plt.show()
```

**Task 5: Histogram Practice** 
1. Create histogram  
2. Analyze distribution  
3. Use random dataset 

```python
#Create Histogram
#Random dataset
data = np.random.randn(1000)

# Create histogram
plt.hist(data, bins=20)

# Add title and labels
plt.title("Histogram of Random Data")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

###### Analyze Distribution
This histogram helps analyze:

Data spread
Frequency of values
Data concentration
Shape of distribution

Most values appear near the center, forming a bell-shaped pattern.
