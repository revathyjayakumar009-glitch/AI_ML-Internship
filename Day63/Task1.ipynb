# 1. Define an AI chatbot.  

An AI chatbot is a software application that uses Artificial Intelligence (AI) and Natural Language Processing (NLP) to understand, interpret, and respond to user queries in a human-like manner. Unlike rule-based chatbots, AI chatbots can learn from data, identify user intent, and provide more accurate and context-aware responses.

Example:

* User: "I want to book a train ticket to Delhi."
* AI Chatbot: "Sure! I can help you book a train ticket to Delhi. Please provide your travel date."

# 2. Explain intent classification.  

Intent classification is the process of identifying the purpose or goal (intent) behind a user's message. It helps a chatbot understand what the user wants so it can provide the correct response.

#### For example:

* User: "Hello!" → Intent: Greeting
* User: "Book a train ticket." → Intent: Book Ticket
* User: "What is the course duration?" → Intent: Course Duration
* User: "Thank you!" → Intent: Thanks
* User: "Bye." → Intent: Goodbye

In AI chatbots, intent classification is typically performed using Natural Language Processing (NLP) and Machine Learning (ML) models that are trained on labeled examples. This allows the chatbot to recognize the intent even when users phrase their requests differently.

# 3. Why is training data important?  

Training data is important because it teaches a machine learning model how to recognize patterns and classify user inputs correctly. It consists of labeled examples that help the chatbot learn different intents and their corresponding responses.

### Importance of Training Data
* Helps the chatbot understand different user intents.
* Improves the accuracy of intent classification.
* Enables the chatbot to handle different ways of asking the same question.
* Reduces incorrect predictions by providing more examples.
* Better quality and larger datasets generally lead to better chatbot performance.

# 4. What is feature extraction?  

Feature extraction is the process of converting text into numerical features that a machine learning model can understand and use for prediction. Since ML algorithms cannot process raw text directly, feature extraction transforms words into numbers while preserving important information.

#### Common Feature Extraction Techniques
* Bag of Words (BoW): Counts how many times each word appears in the text.
* TF-IDF (Term Frequency–Inverse Document Frequency): Assigns higher importance to meaningful words and lower importance to very common words.
* Word Embeddings: Represents words as numerical vectors that capture their meanings and relationships.
