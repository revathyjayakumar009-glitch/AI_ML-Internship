# **Task 2: Regression Practice**
1. Create marks prediction dataset  
2. Predict salary  
3. Predict sales values  

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
#Avoid Warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
#Create marks prdiction dataset
data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 35, 50, 65, 80]
}

df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Marks"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]]).round(2)

print("Predicted Marks:", prediction[0])
```

```python
# Predict Salary
data = {
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [25000, 35000, 45000, 55000, 65000]
}

df = pd.DataFrame(data)

X = df[["Experience"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Salary:", prediction[0])
```

```python
# Predict sales
data = {
    "Month": [1, 2, 3, 4, 5],
    "Sales": [100, 200, 300, 400, 500]
}

df = pd.DataFrame(data)

X = df[["Month"]]
y = df["Sales"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]]).round(2)

print("Predicted Sales:", prediction[0])
```

# **Task 3: Classification Practice** 
1. Pass/fail prediction  
2. Spam/not spam example  
3. Disease prediction example 

```python
#Pass/fail prediction
data = {
    "Hours": [1, 2, 3, 4, 5],
    "Result": [0, 0, 0, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Result"]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Prediction:", prediction[0])
```

```python
#Spam/Not Spam
data = {
    "Length": [20, 50, 100, 150, 200],
    "Spam": [0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Length"]]
y = df["Spam"]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[180]])

print("Prediction:", prediction[0])
```

```python
#Disease prediction example
data = {
    "Fever": [98, 99, 101, 102, 103],
    "Disease": [0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Fever"]]
y = df["Disease"]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[102]])

print("Prediction:", prediction[0])
```

# **Task 4: Identify Problem Type** 
1. House price prediction  
2. Face recognition  
3. Weather forecasting  
4. Fraud detection  
5. Student marks prediction 

# 1. House Price Prediction

Problem Type: Regression <br>

Reason: House price is a numerical value (e.g., ₹50,00,000), so the output is continuous.<br>

Input: House size, location, number of rooms<br>

Output: House price<br>

# 2. Face Recognition

Problem Type: Classification <br>

Reason: The model identifies which person belongs to the face image. The output is a category (person's identity). <br>

Input: Face image <br>

Output: Person A, Person B, Person C, etc. <br>

# 3. Weather Forecasting

Problem Type: Regression <br>

Reason: Weather forecasting often predicts numerical values such as temperature, rainfall, or humidity. <br>

Input: Historical weather data <br>

Output: Temperature (e.g., 32°C) <br>

# 4. Fraud Detection

Problem Type: Classification <br>

Reason: The output belongs to categories: Fraud or Not Fraud. <br>

Input: Transaction details <br>

Output: Fraud / Not Fraud <br>
# 5. Student Marks Prediction

Problem Type: Regression <br>

Reason: Marks are numerical values, so the output is continuous. <br>

Input: Study hours, attendance, assignments <br>

Output: Marks (e.g., 85) <br>

# **Task 6: ML Logic Thinking**
1. Explain why salary prediction is regression  
2. Explain why spam detection is classification  
3. Compare output types 

#### 1. Explain Why Salary Prediction is Regression

Salary prediction is a Regression problem because the output is a numerical value.
The model predicts a salary amount such as ₹50,000 or ₹75,000. Since salary can take many continuous numeric values, it is a Regression problem.
#### 2. Explain Why Spam Detection is Classification
Spam detection is classification because it predicts categories (Spam or Not Spam).
#### 3. Compare Output Types
| Regression Output       | Classification Output |
| ----------------------- | --------------------- |
| Numerical Values        | Categories/Labels     |
| Continuous Output       | Discrete Output       |
| Salary = ₹50,000        | Spam                  |
| House Price = ₹25 Lakhs | Not Spam              |
| Temperature = 30°C      | Pass/Fail             |
| Sales = 1000 Units      | Disease/No Disease    |
