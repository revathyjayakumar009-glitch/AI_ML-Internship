**Task 3: Flatten** 
1. Convert 2D → 1D  
2. Compare flatten vs ravel 

```python
arr = np.array([[1,2,3],
                [5,6,7]])
print(arr)
```

```python
#Convert 2D-->1D
print(arr.flatten())
```

```python
print(arr.ravel())
```

```python
#compare flatten vs ravel
arr = np.array([[1,2,3],
                [4,5,6]])

flat = arr.flatten()
rav = arr.ravel()

flat[0] = 99
flat[3]=100
rav[1] = 88
rav[5]=32

print("Flatten:", flat)
print("Ravel:", rav)
print("Original array:\n", arr)
```

**Task 5: Stacking**
1. Vertical stack  
2. Horizontal stack  
3. Combine arrays 

```python
arr1=np.array([2,4,5,6])
arr2=np.array([11,13,15,17])
```

```python
#Vertical stack
np.vstack((arr1,arr2))
```

```python
#Horizontal stack
np.hstack((arr1,arr2))
```

```python
#Combine arrays
np.stack((arr1,arr2))
```

**Task 6: Splitting** 
1. Split array  
2. Split rows

```python
# Split array
arr = np.array([1,2,3,4,5,6])
res = np.split(arr, 3)
print(res)
```

```python
# Split rows
arr2 = np.array([[1,2],
                 [3,4],
                 [5,6],
                 [7,8]])

res = np.vsplit(arr2, 2)
print(res)
```
