# 1. What is sorting? 

Sorting means arranging data in a specific order, either:

* Ascending order → smallest to largest (1 → 100, A → Z)
* Descending order → largest to smallest (100 → 1, Z → A)<br>

In Pandas, sorting helps organize datasets for better analysis and readability.

# 2. Why sorting is important? 

Sorting is important because it helps us:

**Organize Data Properly**<br>
Arrange data in a clean and readable format.<br>
**Find Highest or Lowest Values Easily**<br>
Example: Top scorer, lowest salary, highest sales, etc.<br>
**Improve Data Analysis**<br>
Makes comparing and understanding datasets much easier.<br>
**Prepare Reports**<br>
Sorted data looks professional in reports and dashboards.<br>
**Work Efficiently with Large Datasets**<br>
In Pandas, sorting helps quickly analyze thousands of rows.

# 3. What is grouping? 

Grouping means combining rows that have the same value in a column and performing operations on those groups.

In Pandas, grouping is mainly done using the **groupby()** function.

Why Grouping is Used?
* To summarize data
* To calculate totals, averages, counts, etc.
* To analyze categories separately* 

# 4. What is aggregation? 

Aggregation means performing calculations on grouped data to get summary results.

In Pandas, aggregation is commonly used with groupby() to calculate:

* Sum
* Mean (average)
* Count
* Maximum
* Minimum

# 5. What is data analysis? 

Data analysis is the process of collecting, organizing, examining, and interpreting data to find useful information and make better decisions.

In Pandas, data analysis helps us understand patterns, trends, and relationships in datasets.

Data Analysis is Used To:
* Find insights from data
* Identify trends and patterns
* Compare values
* Make predictions and decisions
* Generate reports
