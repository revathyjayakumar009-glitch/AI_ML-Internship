# 1. Define Word Embedding. 

Word Embedding is a Natural Language Processing (NLP) technique that converts words into dense numerical vectors while preserving their semantic meaning and relationships. Words with similar meanings are represented by similar vectors, enabling computers to better understand human language.

# 2. What is Word2Vec? 

Word2Vec is a Word Embedding technique developed by researchers at Google. It uses a neural network to learn the meaning of words by analyzing their surrounding words (context) and converts them into dense numerical vectors. Words with similar meanings have similar vector representations.

# 3. Explain CBOW.  

CBOW (Continuous Bag of Words) is one of the two architectures of Word2Vec. It predicts the target word based on its surrounding context words. CBOW is faster to train and works well with large datasets.

#### Example:

Context: "I love ____ learning"

Predicted word: "machine"

Input: Context words
Output: Target word

# 4. Explain Skip-Gram. 

Skip-Gram is one of the two architectures of Word2Vec. It predicts the surrounding context words from a given target word. Skip-Gram performs well with rare words and captures semantic relationships effectively, although it is generally slower to train than CBOW.

### Example:

Target word: machine<br>

Predicted context words:<br>

I<br>
love<br>
learning<br>

Input: Target word<br>
Output: Context words<br>

# 5. Why is Word2Vec better than TF-IDF? 

Word2Vec is better than TF-IDF because it captures the semantic meaning and relationships between words, whereas TF-IDF only measures the importance of words based on their frequency.

### Advantages of Word2Vec over TF-IDF:

* Captures semantic meaning and context.
* Represents words as dense vectors instead of sparse vectors.
* Identifies similarities between related words (e.g., king and queen).
* Improves the performance of NLP tasks such as sentiment analysis, machine translation, and chatbots.
* Produces more meaningful word representations for machine learning models.
