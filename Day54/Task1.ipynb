# 1. What is Bag of Words? 

Bag of Words (BoW) is a text representation technique used in Natural Language Processing (NLP) that converts text into numerical features for Machine Learning models. It creates a vocabulary of unique words from the text and represents each document by counting how many times each word appears, while ignoring grammar and word order.

# 2. Why do we need BoW?  

* Converts text into numerical features
* Prepares text for Machine Learning models
* Helps in text classification
* Used in sentiment analysis, spam detection, and document categorization

# 3. What is a vocabulary? 

A vocabulary is the collection of all unique words found in a set of documents or sentences. In the Bag of Words (BoW) model, each unique word in the vocabulary becomes a feature (column) in the numerical representation.

Example:

Sentences:

"I love AI"
"I love Python"

Vocabulary: ["I", "love", "AI", "Python"]

Each word in the vocabulary is used to create the Bag of Words matrix.

# 4. Explain CountVectorizer. 

CountVectorizer is a class from the sklearn.feature_extraction.text module that converts a collection of text documents into a Bag of Words (BoW) matrix. It builds a vocabulary of unique words and counts how many times each word appears in each document.
### Features:

* Creates a vocabulary of unique words.
* Converts text into numerical vectors.
* Counts the frequency of each word in every document.
* Widely used for text preprocessing in Machine Learning and NLP.

# 5. List BoW limitations. 

### Limitations of Bag of Words (BoW)
* Ignores word order – The sequence of words is not considered.
* Ignores context and meaning – Words with different meanings are treated the same if they have the same spelling.
* Creates sparse matrices – Most values are 0, which can use more memory.
* Large vocabulary size – The number of features increases as more unique words are added.
* Does not capture semantics – It cannot understand relationships between words or synonyms.
