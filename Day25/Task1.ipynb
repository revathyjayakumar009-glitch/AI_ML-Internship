# 2. What is vectorization? 

NumPy Vectorization means performing operations on entire arrays at once instead of using loops.

It helps make code:

1. Faster 
2. Shorter 
3. More efficient 

In NumPy, vectorization allows mathematical operations to run directly on arrays.

```python
#Without Vectorization
numbers = [1, 2, 3, 4]
result = []

for i in numbers:
    result.append(i * 2)

print(result)
```

```python
#with vectorization
import numpy as np

arr = np.array([1, 2, 3, 4])

result = arr * 2

print(result)
```

# 3. Why shape matters in ML?  

In Machine Learning, the shape of data tells us how rows and columns are arranged in an array or matrix.

NumPy arrays use shape to define the dimensions of data.
    Why Shape is Important in ML
**1. Models Expect Correct Input Format**

Machine Learning models require data in a specific shape.

Example:

Rows → samples/data points
Columns → features

If the shape is wrong, the model gives errors.

**2. Helps in Matrix Operations**

ML uses many mathematical operations like:

Matrix multiplication
Dot products
Transformations

These operations work only when shapes are compatible.

**3. Important for Images & Deep Learning**

Images are stored as multidimensional arrays.

Example:

Grayscale image → (height, width)
Color image → (height, width, channels)

Wrong shape can affect training and predictions.

**4. Needed for Reshaping Data**

Sometimes data must be reshaped before training.
**5. Used in Batch Processing**

Deep learning models process data in batches.

# 4. What is normalization? 

Normalization is the process of scaling data into a smaller and common range, usually between 0 and 1.

It helps Machine Learning models work better and faster.
#### **Why Normalization is Important?**

In datasets, values may have very different ranges.

Example:

Age → 18 to 60
Salary → 20,000 to 1,00,000

Large values can dominate smaller values and affect model performance.
Normalization brings all values to a similar scale.


##### Normalization Formula
Xnorm=X−Xmin/Xmax−Xmin
Where:

###### x = original value
###### xmin = minimum value
###### xmax= maximum value

# 5. What is feature extraction? 

Feature extraction is the process of selecting or creating important information (features) from raw data for Machine Learning models.

A feature is an individual measurable property of data.
##### Why Feature Extraction is Important?

Raw data may contain:

1. Unnecessary information
2. Noise
3. Large data size

Feature extraction helps:

1. Reduce complexity
2. Improve model performance
3. Increase training speed
4. Extract useful patterns
