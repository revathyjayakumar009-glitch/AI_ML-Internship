**Task 2: Array Creation** 
1. Create 1D array  
2. Create 2D array  
3. Create 3D array  
4. Print all 

```python
import numpy as np
```

```python
#Create 1D Array
arr1=np.array([1,8,6,9])
print("1D Array:",arr1)

#create 2D Array
arr2=np.array([[1,3,6],[8,10,12]])
print("2D Array:",arr2)
#Create 3D Array
arr3=np.array([[[1,2],[3,4]]])
print("3D Array:",arr3)
```

**Task 3: Properties**

```python
#Print Shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
#print dimensions
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
```

```python
#print Datatypes
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
```

**Task 5: Indexing**

```python
arr = np.array([10, 20, 30, 40])
```

```python
#Accessing first element
print("First element:", arr[0])
```

**Task 6: Reshaping** 
1. Convert 1D → 2D  
2. Convert 2D → 1D

```python
#Convert 1D-> 2D
arr1 = np.array([1, 2, 3, 4, 5, 6])
arr_2d = arr1.reshape(2, 3)
print(arr_2d)
```

```python
#Convert 2D->1D
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
arr_1d = arr2.flatten()
print(arr_1d)
```
