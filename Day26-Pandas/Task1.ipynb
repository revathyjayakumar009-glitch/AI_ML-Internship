# 1. What is Pandas? 

Pandas is an open-source Python library used for:

1. Data analysis
2. Data manipulation
3. Data cleaning
4. Working with tables and datasets

It helps organize data in rows and columns like an Excel sheet.

### Features of Pandas

1. Easy handling of large datasets
2. Fast data processing
3. Supports CSV and Excel files
4. Built on NumPy
5. Widely used in AI/ML and Data Science

# 3. What is Series? 

A Series is a one-dimensional labeled array in Pandas.

It can store:

* Integers
* Strings
* Floats
* Other data types

A Series is similar to a single column in a table.

##### Features of Series

* One-dimensional data structure
*  Stores indexed values
*   Easy data access
*   Fast operations

# 4. What is DataFrame? 

A DataFrame is a two-dimensional data structure in Pandas.

It stores data in:

* Rows
* Columns

A DataFrame is similar to:

* Excel sheet
* Database table
 Features of DataFrame

 Stores tabular data
*  Supports multiple columns
*   Easy data analysis
*   Fast data processing
*   Handles large datasets

# 5. Difference between NumPy and Pandas 

| **Feature**    | **NumPy**                       | **Pandas**                   |
| -------------- | ------------------------------- | ---------------------------- |
| Purpose        | Numerical computing             | Data analysis & manipulation |
| Main Structure | Array                           | Series & DataFrame           |
| Data Type      | Mostly homogeneous              | Heterogeneous supported      |
| Dimension      | Mainly n-dimensional arrays     | Tabular data                 |
| Speed          | Faster for numerical operations | Slightly slower than NumPy   |
| Labels         | No row/column labels            | Supports row & column labels |
| File Handling  | Limited                         | Strong CSV/Excel support     |
| Usage          | Mathematical operations         | Data cleaning & analysis     |
