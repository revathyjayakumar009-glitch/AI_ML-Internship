# 1. What is regression?  

Regression is a type of Machine Learning algorithm used to predict continuous numerical values.

In regression, the output is always a number rather than a category or label. It finds the relationship between input data (features) and output data (target) to make predictions.<br>

**Examples:** <br>

* House price prediction
* Salary prediction
* Temperature prediction
* Sales forecasting<br>
**Example:** <br>
| Experience (Years) | Salary (₹) |
| ------------------ | ---------- |
| 1                  | 20,000     |
| 2                  | 30,000     |
| 3                  | 40,000     |
If a person has 4 years of experience, a regression model can predict the salary as a numerical value.

# 2. What is classification? 

Classification is a type of Machine Learning algorithm used to predict categories or labels.

In classification, the output is a class/category rather than a numerical value. The model learns from labeled data and assigns new data to the correct category.<br>

**Examples:** <br>

* Spam or Not Spam
* Pass or Fail
* Disease or No Disease
* Cat or Dog<br>
**Example:** <br>
| Study Hours | Result |
| ----------- | ------ |
| 1           | Fail   |
| 4           | Pass   |
| 6           | Pass   |
Here, the output (Result) is a category (Pass/Fail), so it is a Classification problem.

# 3. Difference between regression and classification 

| Regression                      | Classification                |
| ------------------------------- | ----------------------------- |
| Predicts numerical values       | Predicts categories or labels |
| Output is continuous            | Output is discrete            |
| Used for quantity prediction    | Used for class prediction     |
| Example: Salary Prediction      | Example: Spam Detection       |
| Example: House Price Prediction | Example: Pass/Fail Prediction |


# 4. What is continuous data? 

Continuous data is data that can take any numerical value within a range, including decimal values.

It is not limited to specific categories and can be measured precisely. <br>

**Examples:**
* Height = 170.5 cm
* Weight = 65.8 kg
* Salary = ₹45,750
* Temperature = 32.7°C

These values can change continuously and can contain decimals.

# 5. What is categorical data? 

ategorical data is data that is divided into groups, categories, or labels rather than numerical values.

It represents different classes or categories and is commonly used in classification problems.<br>

**Examples:**  <br>
* Male / Female
* Pass / Fail
* Spam / Not Spam
* Yes / No

These values represent categories instead of continuous numbers.
