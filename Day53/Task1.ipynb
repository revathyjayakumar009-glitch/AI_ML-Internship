# 1. What is Stemming? 

Stemming is the process of reducing a word to its root or stem by removing prefixes or suffixes. It uses simple rules and does not consider the meaning or grammar of the word.

The resulting stem may or may not be a valid English word.
### Why is Stemming Needed?

In Natural Language Processing (NLP), the same word can appear in different forms. Stemming converts these variations into a common root, helping computers process text more efficiently.

### Example:

play<br>
playing<br>
played<br>
plays<br>

After stemming:<br>

play<br>

All these words are treated as the same term, reducing the size of the vocabulary.

| Original Word | Stem    |
| ------------- | ------- |
| playing       | play    |
| played        | play    |
| studies       | studi   |
| walking       | walk    |
| happily       | happili |
| connections   | connect |


# 2. What is Lemmatization? 

Lemmatization is the process of reducing a word to its base or dictionary form, called the lemma. Unlike stemming, lemmatization considers the meaning, grammar, and part of speech (POS) of a word to produce a valid English word.

The output of lemmatization is always a meaningful dictionary word.
### Why is Lemmatization Needed?

In Natural Language Processing (NLP), the same word can appear in different grammatical forms. Lemmatization converts these forms into a single base word, making text analysis more accurate.

### Example:

run<br>
running<br>
ran<br>
runs<br>

After lemmatization:<br>

run<br>

All these words are treated as the same word.<br>
| Original Word | Lemma |
| ------------- | ----- |
| running       | run   |
| ran           | run   |
| studies       | study |
| better        | good  |
| feet          | foot  |
| children      | child |
| cars          | car   |


# 3. Difference between Stemming and Lemmatization. 

| Feature             | Stemming                                     | Lemmatization                                        |
| ------------------- | -------------------------------------------- | ---------------------------------------------------- |
| Definition          | Removes prefixes or suffixes to get the stem | Converts words to their dictionary base form (lemma) |
| Output              | May not be a valid English word              | Always a valid dictionary word                       |
| Grammar             | Does not consider grammar                    | Considers grammar and context                        |
| Dictionary Required | No                                           | Yes                                                  |
| Accuracy            | Lower                                        | Higher                                               |
| Speed               | Faster                                       | Slower                                               |
| Complexity          | Simple                                       | More complex                                         |
| Processing Method   | Rule-based                                   | Dictionary and grammar-based                         |
| Example             | studies → studi                              | studies → study                                      |
| Best Used For       | Search engines, Information Retrieval        | Chatbots, Machine Translation, Sentiment Analysis    |


# 4. Why is Lemmatization more accurate? 

## 1. Produces Valid Dictionary Words

Lemmatization always returns a valid English word (lemma), whereas stemming may produce incomplete or incorrect words.
| Original Word | Stemming | Lemmatization |
| ------------- | -------- | ------------- |
| studies       | studi    | study         |
| wolves        | wolv     | wolf          |
| children      | children | child         |
## 2. Considers Grammar

Lemmatization analyzes the grammatical role (Part of Speech) of a word before converting it to its base form.
| Word    | Part of Speech | Lemma |
| ------- | -------------- | ----- |
| running | Verb           | run   |
| better  | Adjective      | good  |
| was     | Verb           | be    |
## 3. Understands Different Word Forms

Words with different tenses, plurals, or comparative forms are converted into a common base word.
#### Example
run<br>
running<br>
ran<br>
runs<br>
### After lemmatization:
run
## 4. Preserves Meaning

Lemmatization keeps the original meaning of words.
| Original | Stemming | Lemmatization |
| -------- | -------- | ------------- |
| better   | better   | good          |
| feet     | feet     | foot          |
| mice     | mice     | mouse         |
## Improves NLP Performance

Because lemmatization produces accurate base forms, it helps NLP models understand text more effectively.

##### It improves the performance of:

* Search Engines
* Chatbots
* Sentiment Analysis
* Machine Translation
* Text Classification
* Question Answering Systems
