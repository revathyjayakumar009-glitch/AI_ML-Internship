## **Student Result Dataset**

```python
student_data = pd.DataFrame({
    'Study_Hours': [8, 6, 4, 2, 1],
    'Attendance': [90, 80, 70, 50, 40],
    'Result': ['Pass', 'Pass', 'Pass', 'Fail', 'Fail']
})

#Features and Target
X1 = student_data[['Study_Hours', 'Attendance']]
y1 = student_data['Result']

#Model
student_model = RandomForestClassifier()

#Train
student_model.fit(X1, y1)

#Prediction
student_pred = student_model.predict([[5, 75]])

print("Student Result Prediction:", student_pred[0])
```

```python
#Prediction Practice
student_pred = student_model.predict([[2, 55]])
print("Student Result:", student_pred[0])
```

```python
# n_estimators = 10
student_10 = RandomForestClassifier(n_estimators=10)
student_10.fit(X1, y1)
print("10 Trees:", student_10.predict([[5, 75]])[0])

# n_estimators = 50
student_50 = RandomForestClassifier(n_estimators=50)
student_50.fit(X1, y1)
print("50 Trees:", student_50.predict([[5, 75]])[0])

# n_estimators = 100
student_100 = RandomForestClassifier(n_estimators=100)
student_100.fit(X1, y1)
print("100 Trees:", student_100.predict([[5, 75]])[0])
```

## **Disease Dataset**

```python
disease_data = pd.DataFrame({
    'Age': [25, 30, 22, 40, 28],
    'Fever': [102, 101, 98, 103, 97],
    'Disease': ['Yes', 'Yes', 'No', 'Yes', 'No']
})
#Features and Target
X2 = disease_data[['Age', 'Fever']]
y2 = disease_data['Disease']

#Model
disease_model = RandomForestClassifier()

#Train
disease_model.fit(X2, y2)

# Prediction
disease_pred = disease_model.predict([[35, 102]])

print("Disease Prediction:", disease_pred[0])
```

```python
#Prediction practice
disease_pred = disease_model.predict([[24, 98]])
print("Disease Status:", disease_pred[0])
```

```python
# n_estimators = 10
D_10 = RandomForestClassifier(n_estimators=10)
D_10.fit(X2, y2)
print("10 Trees:", D_10.predict([[35, 102]])[0])

# n_estimators = 50
D_50 = RandomForestClassifier(n_estimators=50)
D_50.fit(X2, y2)
print("50 Trees:", D_50.predict([[35, 102]])[0])

# n_estimators = 100
D_100 = RandomForestClassifier(n_estimators=100)
D_100.fit(X2, y2)
print("100 Trees:", D_100.predict([[35, 102]])[0])
```

## **Loan Approval Dataset**

```python
loan_data = pd.DataFrame({
    'Income': [50000, 40000, 30000, 60000, 25000],
    'Credit_Score': [750, 700, 650, 800, 600],
    'Loan_Approved': ['Yes', 'Yes', 'No', 'Yes', 'No']
})
#Features and Target
X3 = loan_data[['Income', 'Credit_Score']]
y3 = loan_data['Loan_Approved']

#Model
loan_model = RandomForestClassifier()

#Train
loan_model.fit(X3, y3)

#Prediction
loan_pred = loan_model.predict([[45000, 720]])

print("Loan Approval Prediction:", loan_pred[0])
```

```python
#Prediction
loan_pred = loan_model.predict([[28000, 620]])
print("Loan Approval:", loan_pred[0])
```

```python
# n_estimators = 10
loan_10 = RandomForestClassifier(n_estimators=10)
loan_10.fit(X3, y3)
print("10 Trees:", loan_10.predict([[45000, 720]])[0])

# n_estimators = 50
loan_50 = RandomForestClassifier(n_estimators=50)
loan_50.fit(X3, y3)
print("50 Trees:", loan_50.predict([[45000, 720]])[0])

# n_estimators = 100
loan_100 = RandomForestClassifier(n_estimators=100)
loan_100.fit(X3, y3)
print("100 Trees:", loan_100.predict([[45000, 720]])[0])
```

# Comparison of All Three Datasets
| Dataset        | 10 Trees | 50 Trees | 100 Trees |
| -------------- | -------- | -------- | --------- |
| Student Result | Pass     | Pass     | Pass      |
| Disease Status | Yes      | Yes      | Yes       |
| Loan Approval  | Yes      | Yes      | Yes       |
#### Conclusion
* 10 Trees: Fastest training, lower stability.
* 50 Trees: Better stability and accuracy.
* 100 Trees: Most reliable predictions, slightly slower training.
*For these small datasets, the predictions remain the same for all values of n_estimators.
* In larger real-world datasets, increasing the number of trees often improves model performance and reduces overfitting.
