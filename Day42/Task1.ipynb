# 1. What is Naive Bayes?  


Naive Bayes is a supervised machine learning classification algorithm based on Bayes' Theorem. It is called "Naive" because it assumes that all input features are independent of each other.

#### Definition

Naive Bayes predicts the class of a data point by calculating the probability of each class and selecting the class with the highest probability.

#### Key Features
* Used for classification problems
* Based on Bayes' Theorem
* Fast and easy to implement
* Works well with small datasets
* Commonly used in spam detection, sentiment analysis, and disease prediction
#### Example

If a student studies 8 hours and has 90% attendance, Naive Bayes can predict whether the student will Pass or Fail based on probabilities learned from previous data.


# 2. What is Probability? 

2. What is Probability?

Probability is a measure of how likely an event is to occur. It ranges from 0 to 1.

* 0 → Impossible event
* 1 → Certain event
* 0.5 → Equal chance of occurring or not occurring
### Formula
Probability=      Number of Favorable Outcomes /   Total Number of Outcomes

#### Example

When a fair coin is tossed:<br>

Total outcomes = 2 (Head, Tail)<br>
Favorable outcome for Head = 1<br>
P(Head)=1/2<br>
      =0.5<br>

So, the probability of getting a Head is 0.5 (50%).<br>

#### Importance in Machine Learning

Probability helps machine learning algorithms:

* Make predictions
* Handle uncertainty
* Calculate the likelihood of different outcomes
