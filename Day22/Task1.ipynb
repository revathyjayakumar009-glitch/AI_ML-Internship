# 1. What are aggregation functions? 

Aggregation means combining multiple values to produce a single result.
Examples:

###### Sum
###### Mean
###### Maximum
###### Minimum
###### Standard Deviation

Aggregation helps in:

1. Data analysis
2. Finding trends
3. Statistical calculations
4. Machine learning preprocessing

#### Common Aggregation functions
   | Function      | Purpose            |
| ------------- | ------------------ |
| `np.sum()`    | Total sum          |
| `np.mean()`   | Average value      |
| `np.median()` | Middle value       |
| `np.max()`    | Largest value      |
| `np.min()`    | Smallest value     |
| `np.std()`    | Standard deviation |
| `np.var()`    | Variance           |
| `np.argmax()` | Index of maximum   |
| `np.argmin()` | Index of minimum   |


# 2. What is mean? 

Mean is the average value of numbers.

###### Formula

Mean=Number of values/Sum of all values
	​


```python
import numpy as np
```

```python


arr = np.array([10, 20, 30, 40])

print(np.mean(arr))
```

# 3. What is median? 

Median is the middle value in a sorted dataset.

1. If the dataset has an odd number of values → middle number
2. If the dataset has an even number of values → average of two middle numbers

#### Formula Idea

Median=Middle value of sorted data

```python
#Odd number of value

arr = np.array([5, 10, 15, 20, 25])

print(np.median(arr))
```

```python
#Even number of value
arr = np.array([10, 20, 30, 40])

print(np.median(arr))
```

# 4. What is standard deviation? 

Standard Deviation tells us how far the values are spread from the mean (average).

1. Small standard deviation → values are close to the mean
2. Large standard deviation → values are spread out
   Formula



```python


arr = np.array([2, 4, 6, 8, 10])

print(np.std(arr))
```

#### Why Standard Deviation is Important? 🤖

Used in:

1. Data analysis
2. AI & Machine Learning
3. Risk analysis
4. Performance measurement
5. Understanding data consistency

Example:

1. Student marks consistency
2. Stock market analysis
3. Model performance evaluation

# 5. What is axis?  

Axis defines the direction along which an operation is performed in an array.

In a 2D array:

1. axis=0 → Column-wise operation
2. axis=1 → Row-wise operation

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

```python
print(np.sum(arr, axis=0)) #Column wise
```

```python
print(np.sum(arr, axis=1)) #row wise
```
