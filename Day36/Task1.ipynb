# 1. What is train data? 

Train Data (Training Data) is the portion of the dataset used to teach a Machine Learning model.

The model learns:

* Patterns
* Relationships
* Rules<br>
from the training data.

##### Simple Explanation

Just like students learn from textbooks before exams, a machine learning model learns from training data before making predictions.
##### Purpose of Training Data
* To train the model
* To help the model understand patterns
* To build prediction capability

# 2. What is test data? 

Test Data is the portion of the dataset used to check how well a Machine Learning model performs on new unseen data.

It is used after training the model.
##### Simple Explanation

Just like students write exams after studying, a machine learning model is tested using test data after training.
##### Purpose of Test Data
* To evaluate model performance
* To measure prediction accuracy
* To check whether the model works correctly on new data

# 3. What is train-test split? 


Train-Test Split is the process of dividing a dataset into two parts:

1. Training Data
2. Testing Data

It is an important step in Machine Learning.
#### Purpose of Train-Test Split
* Training data is used to teach the model
* Testing data is used to evaluate the model
* Helps check model performance on unseen data

### Simple Example

Suppose a dataset has 100 records:
| Data Type     | Number of Records |
| ------------- | ----------------- |
| Training Data | 80                |
| Testing Data  | 20                |
* 80 records → used for training
* 20 records → used for testing

This is called an 80-20 train-test split.
### Why is it Important?

#### Without train-test split:

* The model may only memorize data
* Evaluation may become incorrect

### With train-test split:

* We can measure real model accuracy
* We can test prediction capability properly

#### Example

from sklearn.model_selection import train_test_split<br>

X_train, X_test, y_train, y_test = train_test_split( <br>
    X, y,<br>
    test_size=0.2,<br>
    random_state=42<br>
)<br>
### Explanation
* test_size=0.2 → 20% data for testing
* random_state=42 → same split every time
### Workflow

Dataset → Split Data → Train Model → Test Model → Evaluate Performance

# 4. Why model evaluation is important? 

In Machine Learning, training a model alone is not enough.
We must check whether the model performs correctly on new unseen data.

This checking process is called Model Evaluation.

### Importance of Model Evaluation
**1. Measures Model Performance**

Model evaluation helps us understand how accurately the model predicts results.

#### Example:

* Student marks prediction
* House price prediction
* Sales forecasting
**2. Checks Accuracy on New Data**<br>

A good model should work well not only on training data but also on unseen test data.<br>
**3. Detects Overfitting and Underfitting**
* Overfitting → Model memorizes training data
* Underfitting → Model fails to learn patterns properly

Evaluation helps identify these problems.<br>
**4. Improves Model Quality**

By evaluating results, we can:

* Tune parameters
* Improve accuracy
Select better algorithms<br>
**5.Ensures Real-World Reliability**

Before using a model in real applications, evaluation ensures the predictions are trustworthy.

### Examples:

* Medical diagnosis systems
* Fraud detection
* Recommendation systems

# 5. What is accuracy? 

Accuracy is a performance metric used in Machine Learning to measure how many predictions made by a model are correct.

### Definition

Accuracy tells us:

**“Out of all predictions, how many were correct?”**
#### Accuracy Formula

**Accuracy=(Correct Predictions/Total Predictions)×100**

**Example**

Suppose a model predicts results for 10 students.<br>

Correct predictions = 8<br>
Wrong predictions = 2<br>

Then:<br>

Accuracy=(8/10)×100=80%<br>

So, the model accuracy is 80%.
### Why Accuracy is Important?
* Measures model performance
* Helps compare models
* Shows prediction correctness
* Used mainly in classification problems
