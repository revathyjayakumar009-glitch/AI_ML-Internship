# 1. What is an Epoch? 

An Epoch is one complete pass of the entire training dataset through the YOLO model during training.

### Example:

Suppose your dataset contains 1,000 images.

* Epoch 1 → YOLO processes all 1,000 images once.
* Epoch 2 → YOLO processes all 1,000 images again.
* Epoch 3 → YOLO processes all 1,000 images a third time.

If you train for 20 epochs, the model sees the entire dataset 20 times.
### Why are Epochs Important?
More epochs allow the model to learn patterns better.
Too few epochs may result in underfitting (poor learning).
Too many epochs may result in overfitting (memorizing training data).

# 2. What is Loss?  


Loss is a measure of how far the model's predictions are from the actual (correct) values during training.

* A high loss means the model is making many errors.
* A low loss means the model's predictions are becoming more accurate.
### In YOLO Training

YOLO typically reports:

* Box Loss – Measures how accurately the bounding boxes match the actual object locations.
* Classification Loss (Cls Loss) – Measures how accurately the model predicts the object class (person, car, dog, etc.).
* DFL Loss (Distribution Focal Loss) – Helps improve bounding box localization accuracy.
### Example
  | Epoch | Box Loss |
| ----- | -------- |
| 1     | 1.85     |
| 5     | 1.20     |
| 10    | 0.80     |
| 20    | 0.40     |
As training progresses, the loss decreases, showing that the model is learning and improving.

# 3. What is Precision? 

Precision measures how many of the objects detected by the YOLO model are actually correct.

### Formula
Precision=  TP / TP + FP
Where:

* TP (True Positives) = Correctly detected objects
* FP (False Positives) = Incorrect detections

### Example

Suppose YOLO detects 100 objects:

90 detections are correct (TP = 90)
10 detections are wrong (FP = 10) <br>
          **Precision=  90 / 90+ 10 = 0.90 = 90%**
### Why is Precision Important?
* High Precision = Fewer false detections.
* Low Precision = More incorrect detections.         

# 4. What is Recall? 

Recall measures how many of the actual objects present in an image are successfully detected by the YOLO model.
### Formula
Recall=  TP / TP + FN
Where:

* TP (True Positives) = Correctly detected objects
* FN (False Negatives) = Objects that were present but not detected
### Example

Suppose an image contains 100 actual objects:

* YOLO correctly detects 85 objects (TP = 85)
* YOLO misses 15 objects (FN = 15)
  Recall= 85 / 85 + 15 = 0.85 = 85%
### Why is Recall Important?
* High Recall = The model detects most of the objects.
* Low Recall = The model misses many objects.  

# 5. What is mAP? 

mAP (Mean Average Precision) is the most important evaluation metric in YOLO object detection. It measures the overall accuracy of the model by combining both Precision and Recall across all object classes.

### What Does mAP Measure?
* How accurately objects are detected.
* How well bounding boxes match the actual objects.
* How effectively the model balances Precision and Recall.
### Common YOLO Metrics
* mAP@50 (mAP50): Uses an IoU threshold of 50%.
* mAP@50-95: Uses multiple IoU thresholds from 50% to 95% and is a stricter evaluation metric.
