# 1. Define a Menu-Driven Chatbot. 

A Menu-Driven Chatbot is a chatbot that interacts with users by displaying a list of predefined options (a menu) and allowing them to select an option instead of typing free-text messages. It responds based on the user's selected choice, making it simple, structured, and easy to use.

# 2. Explain how it works.  

A Menu-Driven Chatbot works by displaying a list of predefined options to the user. The user selects an option by entering the corresponding number or choice. The chatbot then processes the selection, performs the related action or displays the appropriate response, and shows the menu again. This process continues in a loop until the user chooses the Exit option, which ends the conversation.

# 3. List five advantages.  

1. Easy to use – Users can select options without typing long messages.
2. Simple to develop – It requires basic programming logic and predefined responses.
3. Reduces typing errors – Users choose from a menu instead of entering free text.
4. Fast navigation – Users can quickly access the information or service they need.
5. Reliable and consistent – It provides accurate, predefined responses for each menu option.

# 4. List four limitations. 

1. Limited flexibility – It can only respond to the options available in the menu.
2. Cannot understand natural language – It does not interpret free-text questions or complex user inputs.
3. Difficult to scale – Large menus with many options can become confusing and hard to navigate.
4. Less personalized – It provides predefined responses and cannot adapt to individual user needs or context.

# 5. Compare Menu-Driven and Rule-Based Chatbots. 

| **Feature**         | **Menu-Driven Chatbot**                         | **Rule-Based Chatbot**                                        |
| ------------------- | ----------------------------------------------- | ------------------------------------------------------------- |
| **User Input**      | User selects options from a menu.               | User types a message or question.                             |
| **Response Method** | Responds based on the selected menu option.     | Responds based on predefined rules or keywords.               |
| **Ease of Use**     | Very easy, as users choose from listed options. | Requires users to type matching words or phrases.             |
| **Flexibility**     | Limited to available menu choices.              | More flexible than menu-driven but still limited by rules.    |
| **Example**         | Restaurant ordering menu, ATM menu.             | FAQ chatbot, customer support chatbot using keyword matching. |
