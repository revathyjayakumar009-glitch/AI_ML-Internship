**Task 2: Marks Analysis**
1. Find average marks  
2. Find highest marks  
3. Find lowest marks  
4. Find topper 

```python
import numpy as np
```

```python
students = np.array(["Anuja", "Ram", "Meera", "Arjun", "Diya"])
marks = np.array([78, 92, 85, 96, 88])

print("Average marks:",np.mean(marks))
print("Highest marks:",np.max(marks))
print("Lowest marks:",np.min(marks))
topper=np.argmax(marks)
print("Find topper:",students[topper])
```

**Task 3: Image Operations**
1. Increase brightness  
2. Decrease brightness  
3. Normalize image pixels  

```python
image = np.array([[50, 100, 150],
                  [200, 250, 180],
                  [90, 60, 30]])
print(image)
```

```python
#Increase brightness
bright_image = image + 50
print(bright_image)
```

```python
#Dcrease brightness
decrease_image = image-50
print(decrease_image)
```

```python
# Normalize image pixels
normalized_image = image / 255
print(normalized_image)
```

**Task 4: Dataset Filtering**
1. Filter values > 50  
2. Replace low values  
3. Count filtered values 

```python
data = np.array([23, 67, 45, 89, 12, 76, 54, 30, 95])
```

```python
print("Filterd values:",data[data > 50])
data[data<50]=0
print("Replace low values:",data )
filtered_values = data[data > 50]
count_values = np.sum(filtered_values > 50)
print("\nCount of Values Greater Than 50:", count_values)
```

**Task 5: Reshaping**
1. Convert 1D → 2D  
2. Convert 2D → 1D  
3. Create 3D dataset 

```python
#Convert 1D -> 2D
arr1d = np.array([1, 2, 3, 4, 5, 6])
arr2d = arr1d.reshape(2, 3)
print("1D Array:")
print(arr1d)
print("\nConverted to 2D Array:")
print(arr2d)
# Convert 2D → 1D
back_to_1d = arr2d.flatten()
print("\nConverted Back to 1D:")
print(back_to_1d)
# 3. Create 3D dataset
dataset_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("\n3D Dataset:")
print(dataset_3d)
```
