## **Student Result Dataset**

```python
student_data = pd.DataFrame({
    'Study_Hours': [8, 6, 4, 2, 1],
    'Attendance': [90, 80, 70, 50, 40],
    'Result': ['Pass', 'Pass', 'Pass', 'Fail', 'Fail']
})
#Feature and Label
X = student_data[["Study_Hours","Attendance"]] 
y = student_data["Result"] 
 
student_model = SVC() 
 
student_model.fit(X, y) 
 
prediction1 = student_model.predict([[5,75]]) 
 
print("Prediction:", prediction1[0])
```

```python
# prediction practice
student_predictions = student_model.predict([
    [7, 85],   
    [3, 60],   
    [1, 45]   
])

print("Student Predictions:", student_predictions)
```

## **Diesease Dataset**

```python
disease_data = pd.DataFrame({
    'Age': [25, 30, 22, 40, 28],
    'Fever': [102, 101, 98, 103, 97],
    'Disease': ['Yes', 'Yes', 'No', 'Yes', 'No']
})
X_disease = disease_data[['Age', 'Fever']]
y_disease = disease_data['Disease']

disease_model = SVC()
disease_model.fit(X_disease, y_disease)

disease_prediction = disease_model.predict([[35, 102]])

print("Disease Prediction:", disease_prediction[0])
```

```python
# Prediction practice
disease_predictions = disease_model.predict([
    [35, 102],  
    [24, 98],   
    [45, 103]   
])

print("Disease Predictions:", disease_predictions)
```

## **Spam Detection Dataset**

```python
spam_data = pd.DataFrame({
    'Free_Word': [1, 1, 0, 0, 1],
    'Offer_Word': [1, 0, 1, 0, 1],
    'Spam': ['Yes', 'Yes', 'No', 'No', 'Yes']
})
X_spam = spam_data[['Free_Word', 'Offer_Word']]
y_spam = spam_data['Spam']

spam_model = SVC(kernel='linear')
spam_model.fit(X_spam, y_spam)

spam_prediction = spam_model.predict([[1, 0]])

print("Spam Prediction:", spam_prediction[0])
```

```python
# Prediction Practice
spam_predictions = spam_model.predict([
    [1, 1],  
    [0, 0],  
    [0, 1]   
])

print("Spam Predictions:", spam_predictions)
```
