```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Task 2: Correlation Practice**
1. Create dataset  
2. Find correlation matrix  
3. Analyze relationships 

```python
#Create dataset
data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Marks": [35, 50, 65, 80, 95],
    "Sleep_Hours": [8, 7, 6, 5, 4]
}

df = pd.DataFrame(data)

df
```

```python
#Find corelation matrix
cor = df.corr()

cor
```

##### Analyze relationships
**Expected Analysis**
**Study_Hours & Marks**<br>
Strong positive correlation<br>
More study hours → Higher marks<br>
**Study_Hours & Sleep_Hours**<br>
Negative correlation<br>
More study hours → Less sleep<br>
**Marks & Sleep_Hours**<br>
Negative correlation<br>
Higher marks associated with fewer sleep hours in this dataset<br>

**Task 3: Heatmap Practice** 
1. Create heatmap  
2. Show annotations  
3. Analyze strong correlations 

```python
#Create heatmap
sns.heatmap(cor)
```

```python
#Show annotations
sns.heatmap(cor,annot=True, cmap="coolwarm")
```

### Analyse strong correlation
#### Positive Correlation
* Study_Hours and Marks
* Value close to +1
* More study hours lead to higher marks
#### Negative Correlation
* Study_Hours and Sleep_Hours
* Value close to -1
* More study hours reduce sleep hours
#### Strong Correlation
* Values near 1 or -1 indicate strong relationships
#### Weak Correlation
Values near 0 indicate weak relationships

**Task 5: Scatter Plot Practice**
1. Study hours vs marks  
2. Salary vs experience  
3. Temperature vs sales  

```python
study_hours = [1, 2, 3, 4, 5, 6]
marks = [35, 45, 55, 65, 75, 85]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
#Salary Vs Experience
experience = [1, 2, 3, 4, 5, 6]
salary = [25000, 30000, 40000, 10000, 15000, 35000]

plt.scatter(experience, salary)

plt.title("Salary vs Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.show()
```

```python
#Temparature Vs sales
temperature = [20, 22, 25, 28, 30, 35]
sales = [100, 120, 150, 180, 200, 250]

plt.scatter(temperature, sales)

plt.title("Temperature vs Sales")
plt.xlabel("Temperature")
plt.ylabel("Sales")

plt.show()
```

**Task 6: Dataset Analysis** 
1. Student dataset  
2. Employee dataset  
3. Hospital dataset

```python
#Student dataset
student_data = {
    "Name": ["Arun", "Meera", "Rahul", "Anu"],
    "Marks": [85, 92, 78, 88],
    "Study_Hours": [5, 7, 4, 6]
}

students = pd.DataFrame(student_data)

print(students)

print("\nAverage Marks:", students["Marks"].mean())

print("\nHighest Marks:")
print(students[students["Marks"] == students["Marks"].max()])
```

```python
#Employee dataset
employee_data = {
    "Name": ["Aman", "Riya", "John", "Sara"],
    "Salary": [35000, 50000, 45000, 60000],
    "Experience": [2, 5, 4, 6]
}

employees = pd.DataFrame(employee_data)

print(employees)

print("\nAverage Salary:", employees["Salary"].mean())

print("\nEmployee with Highest Salary:")
print(employees[employees["Salary"] == employees["Salary"].max()])
```

```python
#Hospital dataset
hospital_data = {
    "Patient": ["Ravi", "Anjali", "Kiran", "Neha"],
    "Age": [45, 30, 55, 40],
    "Bill": [12000, 8000, 15000, 10000]
}

hospital = pd.DataFrame(hospital_data)

print(hospital)

print("\nAverage Bill Amount:", hospital["Bill"].mean())

print("\nPatient with Highest Bill:")
print(hospital[hospital["Bill"] == hospital["Bill"].max()])
```
