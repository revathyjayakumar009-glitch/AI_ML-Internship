# 1. What is KNN? 

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression tasks. <br>

It works by finding the K nearest data points (neighbors) to a new data point and making a prediction based on those neighbors. <br>

For Classification: Predicts the most common class among the nearest neighbors. <br>
For Regression: Predicts the average value of the nearest neighbors. <br>

# 2. What does K mean in KNN?  

In K-Nearest Neighbors (KNN), K represents the number of nearest neighbors considered when making a prediction.<br>

The algorithm looks at the K closest data points to a new data point and uses them to determine the output.<br>

**Example**<br>

Suppose K = 3 and the three nearest neighbors have the following classes:<br>
| Neighbor | Class |
| -------- | ----- |
| 1        | Pass  |
| 2        | Pass  |
| 3        | Fail  |
Since Pass appears most often (2 out of 3), the prediction is:<br>

Prediction = Pass<br>

* Choosing K <br>
Small K (e.g., K=1): <br>
* More sensitive to noise <br>
Can overfit<br>
Large K (e.g., K=15):<br>
More stable<br>
Can underfit <br>


# 3. Is KNN supervised or unsupervised? 

KNN (K-Nearest Neighbors) is a Supervised Learning algorithm.<br>

It is called supervised because it learns from labeled training data, where the correct output (target) is already known.<br>

# 4. What is a neighbor? 

In KNN (K-Nearest Neighbors), a neighbor is a data point in the training dataset that is closest to the new data point being predicted.<br>

The closeness is usually measured using a distance metric such as Euclidean Distance.<br>

# 5. Advantages of KNN? 

K-Nearest Neighbors (KNN) has several advantages that make it a popular machine learning algorithm:

##### 1. Easy to Understand and Implement
* Simple algorithm based on distance calculations.
* No complex mathematical model is required.
##### 2. No Training Phase
* KNN stores the training data and makes predictions directly.
* This makes the training process very fast.
##### 3. Works for Both Classification and Regression
* Can predict categories (e.g., Pass/Fail).
* Can predict numerical values (e.g., Salary, House Price).
##### 4. Handles Multiple Classes
* Can classify data into more than two categories.
##### 5. Good for Small Datasets
* Performs well when the dataset size is small to medium.
##### 6. Adapts to New Data Easily
* New data can be added without retraining a model.
