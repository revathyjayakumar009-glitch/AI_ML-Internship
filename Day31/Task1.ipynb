# 1. What is subplot? 

A subplot in Matplotlib is a way to display multiple graphs (plots) inside a single figure/window.<br>

It helps compare different datasets or chart types together in one screen.
**Example Layout**

-------------------------
| Plot 1 | Plot 2       |
-------------------------
| Plot 3 | Plot 4       |
-------------------------

**Syntax**<br>

**plt.subplots(rows, columns)**

#### Advantages of Subplots
* Compare multiple charts easily
* Save space
* Create dashboard-style reports
* Visualize related data together


# 2. Why subplots are important? 

Subplots are important because they allow us to display multiple graphs in a single figure, making data analysis easier and more organized.  
#### 1. Easy Comparison

Subplots help compare different datasets or chart types side by side.

**Example**<br>
* Sales vs Profit
* Height vs Weight
* Temperature vs Rainfall
#### 2. Saves Space

Instead of opening many separate graph windows, all charts can be shown in one figure.

#### 3. Better Data Visualization

Subplots create dashboard-style reports that look professional and easy to understand.

#### 4. Improves Analysis

Viewing multiple charts together helps identify:

* Trends
* Patterns
* Relationships
* Differences
#### 5. Used in Real-World Projects

Subplots are commonly used in:

* Data Analytics
* Machine Learning
* Business Reports
* Scientific Research
* Stock Market Analysis

# 3. What is figure size? 

Figure size refers to the width and height of the graph window (figure) in Matplotlib.

It controls how large or small the plot appears.
#### Syntax
plt.figure(figsize=(width, height))
* width → figure width
* height → figure height
* Measured in inches

# 4. What is legend?  

A legend is a key that explains what each plotted line or bar represents. Added with plt.legend() after setting label= on each plot.

# 5. What is histogram? 

A histogram shows the frequency distribution of a dataset — how often each range of values appears. Created with plt.hist(data).
