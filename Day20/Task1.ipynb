# 1. What is indexing? 

Indexing is used to access individual elements in an array.
Array indexing is the same as accessing an array element.
You can access an array element by referring to its index number.
The indexes in NumPy arrays start with 0, meaning that the first element has index 0, and the second has index 1 etc.

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4])

print(arr[0])  #--->  # Accessing first element
print(arr[1])  #--->  #Accessing second element
print(arr[-1]) #--->  #Accessing last element
```

# 2. What is slicing?  

Slicing in python means taking elements from one given index to another given index.

We pass slice instead of index like this: [start:end].

We can also define the step, like this: [start:end:step].

If we don't pass start its considered 0

If we don't pass end its considered length of array in that dimension

If we don't pass step its considered 1

```python
arr = [10, 20, 30, 40, 50]

print(arr[1:4])   
print(arr[:3]) 
print(arr[::2])
```

```python
# Negative slicing
print(arr[-3:-1])
```

```python
# Slicing 2D Array
arr = [[1, 2, 3],
       [4, 5, 6],
       [7, 8, 9]]

print(arr[0:2][0:2])
```

#  3. Difference between indexing and slicing 

| Feature            | **Indexing**                | **Slicing**                    |
| ------------------ | --------------------------- | ------------------------------ |
| 📌 Meaning         | Access a **single element** | Access a **range of elements** |
| 🔢 Syntax          | `arr[index]`                | `arr[start:stop:step]`         |
| 📤 Output          | Single value                | List/array (multiple values)   |
| 🚫 Stop Rule       | Not applicable              | **Stop index is excluded**     |
| 🔁 Multiple values | ❌ No                        | ✅ Yes                       |
| 📚 Use Case        | Get one specific item       | Get subset of data             |


# 4. What is boolean indexing?  

Boolean indexing is a method used to filter elements from a list or array based on a condition.
It uses True/False (Boolean values) to select data.

**How It Works**
######  A condition is applied to the array
###### It returns True/False for each element
###### Only elements with True are selected

```python
arr = np.array([10, 25, 30, 15, 50])

print(arr > 20)
```

```python
print(arr[arr > 20 ])
```

# 5. Why indexing is important in ML? 

Indexing isn’t just a basic concept—it’s essential for handling and manipulating data efficiently in Machine Learning.

##### 1. Access Specific Data Quickly

In ML, datasets can be very large. Indexing helps you pick specific rows or values instantly.

##### 2. Feature Selection

You often need to select specific columns (features) from a dataset.
##### 3. Data Preprocessing

Before training a model, data must be cleaned and prepared.

Indexing helps to:

Remove unwanted values
Replace missing data
Select required subsets
##### 4. Filtering Data (with Conditions)

Using indexing + conditions (Boolean indexing), you can filter useful data.
##### 5. Train-Test Splitting

You need to divide data into training and testing sets.
##### 7. Faster Computation

Efficient indexing avoids unnecessary loops, making ML models faster and more optimized.
