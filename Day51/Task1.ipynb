# 1. What is NLP? 

Natural Language Processing (NLP) is a branch of Artificial Intelligence (AI) that enables computers to understand, interpret, analyze, and generate human language.

NLP combines Computer Science, Linguistics, and Machine Learning to help computers communicate with humans in a natural and meaningful way.

### Definition

Natural Language Processing (NLP) is the technology that allows computers to process and understand human languages such as English, Malayalam, Hindi, and many others.
### Examples of NLP Applications
💬 Chatbots (e.g., ChatGPT) <br>
🎙️ Voice Assistants (Siri, Alexa, Google Assistant)<br>
🌐 Language Translation (Google Translate)<br>
📧 Spam Email Detection<br>
😊 Sentiment Analysis<br>
🔍 Search Engines<br>
📝 Text Summarization<br>

# 2. Why do we need NLP? 

Computers understand only binary data (0s and 1s), while humans communicate using words, sentences, and natural language. This difference makes it difficult for computers to understand human communication directly.

Natural Language Processing (NLP) acts as a bridge between humans and computers by converting human language into a format that computers can understand and process.
### Why is NLP Important?
* Enables computers to understand human language.
* Makes human-computer interaction more natural.
* Helps analyze large amounts of text data quickly.
* Automates language-based tasks such as translation, summarization, and question answering.

# 3. Explain three NLP applications. 

## 1. Chatbots and Virtual Assistants 

NLP enables chatbots and virtual assistants to understand user questions and provide relevant responses in natural language.

### Examples:

* ChatGPT
* Siri
* Alexa
* Google Assistant
## 2. Language Translation 

NLP helps translate text or speech from one language to another while preserving the meaning.

### Examples:

* Google Translate
* Microsoft Translator

### Example:

* English: "Good Morning"
* Malayalam: "സുപ്രഭാതം"
## 3. Sentiment Analysis 😊

NLP analyzes text to identify the emotion or opinion expressed, such as positive, negative, or neutral. It is widely used to analyze customer feedback and social media posts.

### Examples:

* "This product is amazing!" → Positive 😊
* "The service was very poor." → Negative 😞
* "The product arrived today." → Neutral 😐

# 4. Difference between NLP and Computer Vision.  

| **Natural Language Processing (NLP)**                                        | **Computer Vision (CV)**                                                                      |
| ---------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------- |
| Works with **text and language data**.                                       | Works with **images and videos**.                                                             |
| Helps computers understand, analyze, and generate human language.            | Helps computers understand and interpret visual information.                                  |
| Input is **text or speech**.                                                 | Input is **images or video frames**.                                                          |
| Output includes translated text, chatbot responses, sentiment analysis, etc. | Output includes object detection, image classification, face recognition, etc.                |
| Examples: ChatGPT, Google Translate, spam detection.                         | Examples: Face recognition, self-driving cars, medical image analysis, YOLO object detection. |


# 5. What is text data? 

Text data is any information that is written using words, letters, numbers, or symbols. It is the primary type of data used in Natural Language Processing (NLP).
NLP processes text data to understand its meaning, analyze it, and generate meaningful responses.
### Definition

Text data is written information in the form of words, sentences, or paragraphs that can be processed and analyzed by computers using NLP techniques.
### Examples of Text Data
* Hello
* How are you?
* AI is amazing.
* Welcome to Python.
* Machine Learning is the future.
* Customer reviews and emails.
