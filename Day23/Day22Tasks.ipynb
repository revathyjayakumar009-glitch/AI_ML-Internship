**Task 3: Array Broadcasting** 
1. Add two compatible arrays  
2. Multiply arrays  
3. Observe shape changes  

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])
```

```python
print("Add two compatible array:",a+b)
print("Multiply arrays:",a*b)
print("Observe shape changes")
print(a.shape)
print(b.shape)
```

**Task 4: Shape Practice** 
1. Print shapes  
2. Check compatible shapes  
3. Try incompatible shapes 

```python
#print shapes

a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a.shape)
print(b.shape)
```

```python
#Check compatible shapes


b = np.array([10,20,30])

print(a + b)   # compatible because dimensions are equal
```

```python
# try incompatibe shapes
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20])

print(a + b)
```

**Task 5: 2D Broadcasting** 
1. Add scalar to matrix  
2. Multiply matrix  
3. Normalize matrix values 

```python
#Add scalar to matrix
matrix = np.array([[1,2,3],
                   [4,5,6]])

print(matrix + 10)
```

```python
#Multyply matrix
print(matrix * 5)
```
