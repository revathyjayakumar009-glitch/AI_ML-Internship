# 1. Define a chatbot. 

A chatbot is a software application that simulates human conversation through text or voice. It uses predefined rules or Artificial Intelligence (AI) to understand user queries and provide relevant responses or perform tasks automatically.

# 2. Explain chatbot architecture.  

### 1. User Interface (UI):
* The platform where users interact with the chatbot (website, mobile app, WhatsApp, etc.).
### 2. Input Processing:
* Receives the user's text or voice input.
* Cleans and prepares the input for analysis.
### 3. Natural Language Processing (NLP):
* Understands the meaning and intent of the user's message.
* Identifies keywords, entities, and context.
### 4. Dialogue Manager:
* Decides the appropriate response based on the user's intent and conversation flow.
### 5. Knowledge Base / Database:
* Stores predefined responses, FAQs, or retrieves information from databases and APIs.
### 6. Response Generation:
* Creates and sends the most appropriate reply back to the user.
### Simple Flow

## **User → User Interface → Input Processing → NLP → Dialogue Manager → Knowledge Base/API → Response Generation → User**

This architecture enables a chatbot to understand user requests and provide relevant, accurate responses efficiently.

# 3. What is NLP in chatbots?  

Natural Language Processing (NLP) is a branch of Artificial Intelligence (AI) that enables chatbots to understand, interpret, and generate human language.

In chatbots, NLP helps the system:

* Understand the user's intent.
* Identify important keywords and entities.
* Process text or voice input.
* Generate accurate and natural responses.
* Support conversations in multiple languages.

Example:

* User: "What's the weather today?"
* Chatbot (using NLP): Understands that the user is asking for weather information and provides the current forecast.

# 4. Differentiate Rule-Based and AI-Based chatbots. 

| **Rule-Based Chatbots**                         | **AI-Based Chatbots**                                             |
| ----------------------------------------------- | ----------------------------------------------------------------- |
| Work using predefined rules and decision trees. | Use Artificial Intelligence (AI), Machine Learning (ML), and NLP. |
| Respond only to specific keywords or commands.  | Understand natural language and user intent.                      |
| Cannot learn from interactions.                 | Can learn and improve over time.                                  |
| Best for simple and repetitive tasks.           | Suitable for complex and dynamic conversations.                   |
| Easy to develop and maintain.                   | More complex to develop and requires training data.               |
| **Example:** FAQ chatbot.                       | **Example:** ChatGPT, Google Gemini, Microsoft Copilot.           |


# 5. List five real-world applications of chatbots. 

1. Customer Support – Answer customer queries and provide 24/7 assistance.
2. Healthcare – Schedule appointments, provide health information, and send medication reminders.
3. Banking & Finance – Check account balances, transfer funds, and answer banking-related questions.
4. E-commerce – Recommend products, track orders, and assist with returns or exchanges.
5. Education – Help students with learning materials, answer questions, and provide personalized tutoring.
