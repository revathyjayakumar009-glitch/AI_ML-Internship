# **Practical Task 1** 
Create a chatbot that responds to: <br>
• Hello  
• Hi  
• Bye  
• Your Name  

```python
print("🤖 Welcome to the Rule-Based Chatbot!")
print("Type 'bye' to exit.\n")

while True:
    user = input("You: ").lower()

    if user == "hello":
        print("Bot: Hello! How can I help you?")
    elif user == "hi":
        print("Bot: Hi! Nice to meet you.")
    elif user == "your name":
        print("Bot: My name is Python Chatbot.")
    elif user == "bye":
        print("Bot: Goodbye! Have a great day!")
        break
    else:
        print("Bot: Sorry, I don't understand that.")
```

# **Practical Task 2** 
Create a College Enquiry Chatbot with at least five questions.

```python
print("🎓 Welcome to the College Enquiry Chatbot!")
print("Ask your questions or type 'bye' to exit.\n")

while True:
    user = input("You: ").lower()

    if "admission" in user:
        print("Bot: Admissions are open. Please visit the admission office or apply online.")
    elif "courses" in user:
        print("Bot: We offer BCA, B.Sc, B.Com, BA, M.Com, MCA, and MBA programs.")
    elif "fees" in user:
        print("Bot: Fee details are available on the college website or at the accounts office.")
    elif "hostel" in user:
        print("Bot: Yes, separate hostel facilities are available for boys and girls.")
    elif "library" in user:
        print("Bot: Our library has thousands of books, journals, and digital resources.")
    elif "bye" in user:
        print("Bot: Thank you for visiting. Have a nice day!")
        break
    else:
        print("Bot: Sorry, I don't have information about that.")
```

# **Practical Task 4** 
Add keyword matching so that the chatbot responds even if the user types a longer <br>
sentence.<br> 
Example: <br>
Hello there <br>
↓ <br>
Hello!

```python
print("🤖 Welcome to the College Enquiry Chatbot!")
print('Type "bye" to exit.\n')

responses = {
    "hello": "Hello! How can I help you?",
    "hi": "Hi! Welcome to our college.",
    "admission": "Admissions are open. Please visit the admission office or apply online.",
    "fees": "Fee details are available on the college website or at the accounts office.",
    "courses": "We offer BCA, B.Sc, B.Com, BA, M.Com, MCA, and MBA programs.",
    "hostel": "Yes, separate hostel facilities are available for boys and girls.",
    "library": "Our library has thousands of books, journals, and digital resources.",
    "bye": "Thank you for visiting. Have a great day!"
}

while True:
    user = input("You: ").lower()

    found = False

    for keyword, response in responses.items():
        if keyword in user:
            print("Bot:", response)
            found = True

            if keyword == "bye":
                break

            break

    if found and "bye" in user:
        break

    if not found:
        print("Bot: Sorry, I don't understand your question.")
```
