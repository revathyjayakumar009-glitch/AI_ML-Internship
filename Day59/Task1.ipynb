# 1. Define Rule-Based Chatbot. 

A Rule-Based Chatbot is a chatbot that responds to user queries using predefined rules, keywords, and if-else conditions. It does not learn from data or understand context; it only provides responses that have been programmed in advance.

# 2. Explain the working of a Rule-Based Chatbot. 

A Rule-Based Chatbot works by following predefined rules and matching user input with programmed keywords or conditions. Its working process is:

1. The user enters a message.
2. The chatbot receives and processes the input.
3. It checks the input against predefined rules using if-elif-else statements or keyword matching.
4. If a matching rule is found, the chatbot returns the corresponding response.
5. If no rule matches, it displays a default response such as "Sorry, I don't understand your question."

A rule-based chatbot only responds to questions it has been programmed to handle and cannot learn or understand context like AI-based chatbots.

# 3. What is the purpose of lower()?  

The lower() method in Python converts all the characters in a string to lowercase. It helps make user input case-insensitive, so the chatbot can recognize words regardless of whether the user types uppercase or lowercase letters.

# 4. Why is a dictionary useful in chatbots? 

A dictionary is useful in chatbots because it stores keywords as keys and their corresponding responses as values. This makes the chatbot code simpler, faster, and easier to manage compared to using many if-elif-else statements.

# 5. List the advantages and limitations of Rule-Based Chatbots.  

5. List the advantages and limitations of Rule-Based Chatbots.

### Advantages
1. Easy to build and implement.
2. No training data is required.
3. Provides fast and consistent responses.
4. Suitable for FAQs and simple customer support.
5. Low development and maintenance cost.
### Limitations
1. Cannot learn from user interactions.
2. Understands only predefined keywords and rules.
3. Cannot handle complex or unexpected questions.
4. Does not understand context or natural language.
5. Limited flexibility compared to AI-based chatbots.
