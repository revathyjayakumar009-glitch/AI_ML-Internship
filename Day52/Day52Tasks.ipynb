# **Practical Task 3**
Remove punctuation from: <br>
Python!!! is Awesome???<br>

```python
import string
text = "Python!!! is Awesome???"
clean_text = text.translate( 
str.maketrans('', '', string.punctuation) 
) 
print(clean_text)
```

# **Practical Task 5**
Clean this sentence: 
AI 2025 is AMAZING!!!  

```python
import re
text = "AI 2025 is AMAZING!!!      "
text = text.lower() 
text = re.sub(r'[^a-zA-Z\s]', '', text) 
print(text)
```
