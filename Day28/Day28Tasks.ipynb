**Task 2: Missing Data Practice** 
1. Create dataset with missing values  
2. Detect missing values  
3. Count missing values  

```python
import pandas as pd
import numpy as np
```

```python

data = {
    "Name": ["Anu", "Rahul", "Meera", "Arjun", "Diya",
             "Kiran", "Sneha", "Akhil", "Niya", "Vimal"],

    "Marks": [85, np.nan, 90, 78, np.nan,
              67, np.nan, 88, 92, np.nan],

    "Age": [20, 21, np.nan, 22, 20,
            np.nan, 23, 21, np.nan, 24],

    "City": ["Kochi", "Calicut", np.nan, "Trivandrum", "Kannur",
             np.nan, "Kollam", "Kochi", np.nan, "Malappuram"]
}

df = pd.DataFrame(data)

df
```

```python
#Detect Missing values
df.isnull()
```

```python
#Count Missing Values
df.isnull().sum()  
```

**Task 3: Remove Missing Data** 
1. Drop rows with missing values  
2. Drop columns with missing values  

```python
#Drop rows with missing values
df.dropna()
```

```python
#Drop columns with missing values
df.dropna(axis=1)
```

**Task 4: Fill Missing Data**
1. Fill with text  
2. Fill with mean value  
3. Fill numeric columns  

```python
#Fill with text
df["City"]=df["City"].fillna("Unknown")
```

```python
df
```

```python
#Fill with mean value
df["Marks"] = df["Marks"].fillna(df["Marks"].mean().round(2))
```

```python
df
```

```python
#Fill numeric column
df["Age"] = df["Age"].fillna(df["Age"].mean()).round(2)
df
```

**Task 5: Duplicate Handling** 
1. Create duplicate rows  
2. Detect duplicates
3. Remove duplicates   

```python
#create duplicate rows
data = {
    "Name": ["Anu", "Rahul", "Meera", "Anu", "Diya",
             "Rahul", "Sneha", "Akhil"],

    "Marks": [85, 78, 90, 85, 67,
              78, 88, 92],

    "City": ["Kochi", "Calicut", "Trivandrum", "Kochi",
             "Kannur", "Calicut", "Kollam", "Kochi"]
}

df1 = pd.DataFrame(data)

print(df1)
```

```python
#Detect duplicate
df1.duplicated()
```

```python
#Remove duplicates
df1.drop_duplicates()
```
