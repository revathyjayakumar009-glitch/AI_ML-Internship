# 1. What is NumPy random module? 

The NumPy random module is used to generate random numbers, random arrays, and simulated datasets efficiently using NumPy.

It is mainly used in:

 1. AI & Machine Learning
 2. Data Science
 3. Simulations
 4. Testing and experiments

The module is available inside NumPy as:

```python
import numpy as np
```

```python
np.random
```

### Common Functions
| Function              | Purpose                                |
| --------------------- | -------------------------------------- |
| `np.random.rand()`    | Random decimal numbers                 |
| `np.random.randint()` | Random integers                        |
| `np.random.randn()`   | Random values from normal distribution |
| `np.random.choice()`  | Random selection                       |
| `np.random.shuffle()` | Shuffle array                          |

# 3. What is seed? 

A seed is a starting value used to generate random numbers.When the same seed value is used, NumPy produces the same random output every time.
This helps make results reproducible and consistent.

**Why is Seed Important?**

1. Reproduces the same results
2.  Useful in Machine Learning experiments
3.  Helps in debugging
4.  Ensures consistent testing

```python
np.random.seed(42)

print(np.random.randint(1, 10, 5))
```

# 4. Difference between randint and rand  


| Feature     | `randint()`          | `rand()`              |
| ----------- | -------------------- | --------------------- |
| Output Type | Integer              | Float                 |
| Value Range | Custom range         | 0 to 1                |
| Used For    | Random whole numbers | Random decimal values |
