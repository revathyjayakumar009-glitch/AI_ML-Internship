```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

# 1. What is vectorization? 

Vectorization means performing operations on entire arrays at once without using explicit loops.
 Instead of processing element-by-element (like in normal Python), NumPy processes whole arrays in a single step.

# Without Vectorization (Using Loops)

```python
a = [1, 2, 3, 4]
b = [5, 6, 7, 8]

result = []

for i in range(len(a)):
    result.append(a[i] + b[i])

print(result)
```

###### Slower
###### More code
###### Not efficient

# With Vectorization (NumPy Way)

```python
result = a + b
print(result)
```

###### Faster
###### Cleaner
###### Efficient

# 2. Difference between list and NumPy operations 

| Feature              | Python List    | NumPy Array                    |
| -------------------- | ---------------- | --------------------------- |
| Operation Type       | Not element-wise | Element-wise                |
| Addition (`+`)       | Concatenation    | Mathematical addition       |
| Multiplication (`*`) | Repetition       | Element-wise multiplication |
| Speed                | Slower           | Faster (vectorized)         |
| Memory               | Less efficient   | More efficient              |
| Code Style           | Requires loops   | No loops needed             |
| Use Case             | General purpose  | Data science, ML            |

# 4. What is axis? 

In NumPy, an axis refers to the direction along which operations are performed in an array.

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
```

Axis = 0 (Columns ⬇️)
Operation goes down the rows
Works column-wise

```python
np.sum(a, axis=0)
```

Axis = 1 (Rows ➡️)
Operation goes across columns
Works row-wise

```python
np.sum(a, axis=1)
```

# 5. Why NumPy is faster? 

NumPy is much faster than regular Python lists because of how it handles data and computations internally.

#### Main Reasons

###### Vectorization (No Loops)


NumPy performs operations on entire arrays at once


Avoids slow Python loops

Faster execution 

###### Written in C (Low-Level Speed)


Core NumPy operations are implemented in C language


Closer to hardware → faster performance



###### Contiguous Memory Storage


Data is stored in continuous blocks of memory


Improves access speed and cache efficiency



###### Less Memory Usage


NumPy arrays use fixed data types


Python lists store extra metadata for each element


 More compact & efficient

###### Broadcasting


Performs operations between arrays of different shapes


No need for manual looping



###### SIMD Optimization


Uses Single Instruction, Multiple Data


Performs the same operation on multiple values simultaneously
