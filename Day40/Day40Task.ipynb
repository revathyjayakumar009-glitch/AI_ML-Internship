# 1. Create Student result dataset

```python
student_data = pd.DataFrame({
    'Study_Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Attendance': [60, 65, 70, 75, 80, 85, 90, 95],
    'Result': ['Fail', 'Fail', 'Fail', 'Pass', 'Pass', 'Pass', 'Pass', 'Pass']
})

#student_data

# Features and Target
X_student = student_data[['Study_Hours', 'Attendance']]
y_student = student_data['Result']

# Create Model
student_model = DecisionTreeClassifier(random_state=42)

# Train Model
student_model.fit(X_student, y_student)

# Prediction
student_prediction = student_model.predict([[5, 80]])

print("Student Result Prediction:", student_prediction[0])
```

```python
#Prediction practice
student_prediction = student_model.predict([[6, 85]])

print("Student Result:", student_prediction[0])
```

# 2.Create loan approval dataset


```python
loan_data = pd.DataFrame({
    'Income': [20000, 30000, 40000, 50000, 60000, 70000, 80000, 90000],
    'Credit_Score': [550, 600, 620, 680, 720, 750, 800, 850],
    'Loan_Approved': ['No', 'No', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes']
})

#loan_data
# Features and Target
X_loan = loan_data[['Income', 'Credit_Score']]
y_loan = loan_data['Loan_Approved']

# Create Model
loan_model = DecisionTreeClassifier(random_state=42)

# Train Model
loan_model.fit(X_loan, y_loan)

# Prediction
loan_prediction = loan_model.predict([[65000, 740]])

print("Loan Approval Prediction:", loan_prediction[0])
```

```python
# Predict loan approval
loan_prediction = loan_model.predict([[65000, 740]])

print("Loan Approval:", loan_prediction[0])
```

# 3. Create disease prediction dataset

```python
disease_data = pd.DataFrame({
    'Temperature': [98, 99, 100, 101, 102, 103, 104, 105],
    'Cough': [0, 0, 0, 1, 1, 1, 1, 1],
    'Disease': ['No', 'No', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes']
})

#disease_data
# Features and Target
X_disease = disease_data[['Temperature', 'Cough']]
y_disease = disease_data['Disease']

# Create Model
disease_model = DecisionTreeClassifier(random_state=42)

# Train Model
disease_model.fit(X_disease, y_disease)

# Prediction
disease_prediction = disease_model.predict([[103, 1]])

print("Disease Prediction:", disease_prediction[0])
```

```python
#Predict Disease status
disease_prediction = disease_model.predict([[103, 1]])
print("Disease Status:", disease_prediction[0])
```
