```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": [
        "Anu", "Rahul", "Meera", "Arjun", "Diya",
        "Kiran", "Sneha", "Vivek", "Akhil", "Neha"
    ],

    "Department": [
        "HR", "IT", "HR", "Finance", "IT",
        "Marketing", "Finance", "IT", "HR", "Marketing"
    ],

    "City": [
        "Kochi", "Calicut", "Kochi", "Trivandrum", "Kannur",
        "Kochi", "Calicut", "Trivandrum", "Kannur", "Kochi"
    ],

    "Category": [
        "A", "B", "A", "C", "B",
        "A", "C", "B", "A", "C"
    ],

    "Salary": [
        30000, 50000, 35000, 45000, 52000,
        40000, 47000, 55000, 32000, 42000
    ]
}

df = pd.DataFrame(data)
```

**Task 2: Sorting Practice** 
1. Sort salary ascending  
2. Sort salary descending  
3. Sort by multiple columns 

```python
#Sort salary ascending
df.sort_values("Salary")
```

```python
#Sort Salary Decending
df.sort_values("Salary",ascending=False)
```

```python
#Sort by multiple Columns
df.sort_values(["Department","Salary"])
```

**Task 3: Grouping Practice**
1. Group by department  
2. Group by city  
3. Group by category 

```python
#Group by department
df.groupby("Department")["Salary"].mean().round(2)
```

```python
#Group by City
df.groupby("City")["Salary"].sum()
```

```python
#Group by Category
df.groupby("Category")["Salary"].count()
```

**Task 4: Aggregation Practice** 
1. Find average salary

```python
#Find Average Salary
print("Average:",df["Salary"].mean())
```

```python
#Find total Salary
print("Sum:",df["Salary"].sum())
```
