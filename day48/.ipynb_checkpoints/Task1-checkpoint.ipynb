# 1. What is Custom Object Detection? 

Custom Object Detection is the process of training an object detection model to identify and locate user-defined objects that are not necessarily part of the model's original training classes.

Unlike a standard pretrained model that can detect common objects (e.g., people, cars, dogs), a custom object detector is trained on your own labeled dataset so it can recognize specific objects relevant to your application.

### How It Works

A custom object detection model performs two tasks simultaneously:

* Classification – Determines what the object is.
* Localization – Draws a bounding box around the object.

### Why Use Custom Object Detection?

Pretrained models may not recognize specialized objects such as:

* Safety helmets
* Industrial defects
* Medical instruments
* Plant diseases
* Company-specific products
* Wildlife species

By training on your own dataset, the model learns these custom categories.

# 2. What is Annotation? 

Annotation is the process of labeling data so that a machine learning model can learn from it. In object detection, annotation involves drawing bounding boxes around objects in images and assigning the correct class label to each object.

### Example

Suppose you have an image containing a person wearing a helmet.

During annotation, you would:

* Draw a box around the helmet and label it "helmet"
* Draw a box around the person and label it "person"
### Why is Annotation Important?

Annotations act as the ground truth for training. The model learns:

* What the object is (classification)
* Where the object is located (localization)

Without accurate annotations, the model cannot learn to detect objects correctly.

# 3. What is data.yaml? 

data.yaml is a dataset configuration file used by YOLO to describe your dataset. It tells the model:

* Where the training images are located
* Where the validation images are located
* How many classes exist
* The names of those classes

Without data.yaml, YOLO does not know how to access or interpret your dataset.
### Example
path: dataset <br>

train: images/train <br>
val: images/val  <br>

names: <br>
  0: helmet  <br>
  1: person  <br>
  

# 4. What is an Epoch? 

An epoch is one complete pass of the entire training dataset through the machine learning model during training.

In other words, during one epoch, the model sees every training image once and updates its weights based on the errors it makes.

# 5. What is best.pt?  

best.pt is the best-performing trained YOLO model file saved during training. It contains the model's learned weights and parameters from the epoch that achieved the highest validation performance.

The .pt extension stands for a PyTorch model file.
