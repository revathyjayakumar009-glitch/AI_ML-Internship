# 2. What is ML model? 

A Machine Learning (ML) model is a program or algorithm that learns patterns from data and makes predictions or
decisions without being explicitly programmed for every task.<br>
An ML model learns from past data and uses that knowledge to predict future outcomes.<br>
How an ML Model Works<br>
**Step 1: Training**

The model is trained using data.

Example:

Study hours → Marks scored

The model learns the relationship between input and output.<br>
**Step 2: Prediction**

After training, the model predicts results for new data.

Example:

If a student studies for 6 hours, the model predicts marks.
#### Components of an ML Model
| Component       | Meaning                       |
| --------------- | ----------------------------- |
| Input (Feature) | Data given to model           |
| Output (Label)  | Result predicted by model     |
| Training Data   | Data used for learning        |
| Algorithm       | Method used to learn patterns |

# 3. What is feature? # 

In Machine Learning, a feature is an input variable or attribute used by the model to make predictions.

##### Simple Definition

A feature is the information given to the ML model for learning and prediction.
#### Example
##### Student Marks Prediction
| Study Hours | Marks |
| ----------- | ----- |
| 2           | 30    |
| 4           | 50    |
| 6           | 80    |
Here:

Study Hours → Feature (Input)
Marks → Label/Target (Output)

The ML model uses the feature to predict marks.

##### Features are Also Called
* Input variables
* Independent variables
* Predictors

# 4. What is label? 

In Machine Learning, a label is the output or target value that the model tries to predict.

Simple Definition

A label is the correct answer the ML model learns from during training.
#### Example
##### Student Marks Prediction
| Study Hours | Marks |
| ----------- | ----- |
| 2           | 30    |
| 4           | 50    |
| 6           | 80    |
Here:

Study Hours → Feature (Input)
Marks → Label/Target (Output)
The model learns how study hours affect marks.
##### Labels are Also Called
* Target variable
* Output variable
* Dependent variable
  

# 5. What is prediction? 

In Machine Learning, prediction is the process of using a trained model to estimate or guess the output for new input data.

Simple Definition

Prediction means the ML model gives an output based on learned patterns from training data.
#### Example
##### Student Marks Prediction
| Study Hours | Marks |
| ----------- | ----- |
| 2           | 30    |
| 4           | 50    |
| 6           | 80    |
After training, if we give:

Study Hours = 5

The model may predict:

Marks = 70

This output is called a prediction.
### How Prediction Works
#### Step 1: Train the Model

The model learns from existing data.

#### Step 2: Give New Input

New data is provided to the model.

#### Step 3: Model Predicts Output

The model generates a result based on learned patterns.
